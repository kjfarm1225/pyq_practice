```python
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
df_plan = pd.DataFrame(
    [
        ["SUS", "J69P", 50000],
        ["SUS", "310D", 30000],
        ["WSE", "310D", 45000],
        ["GSE", "J69P", 32000],
        ["MELMB", "310D", 60000],
        ["ISE", "J69P", 20000],
    ],
    columns=[
        "site",
        "model",
        "plan_qty",
    ],
)
df_plan
```

```python
#cell3
df_defect = pd.DataFrame(
    [
        ["SUS", "J69P", "コイル", 3],
        ["SUS", "J69P", "Buck-IC", 1],
        ["SUS", "J69P", "コイル", 2],
        ["SUS", "310D", "NTF", 1],
        ["WSE", "310D", "コイル", 2],
        ["WSE", "310D", "NTF", 0],
        ["GSE", "J69P", "コイル", 4],
        ["GSE", "J69P", "Buck-IC", 2],
        ["MELMB", "310D", "コイル", 5],
        ["MELMB", "310D", "NTF", 1], 
    ],
    columns=[
        "site",
        "model",
        "defect_category",
        "defect_qty",
    ],
)
df_defect
```

```python
df_work = df_defect[
    [
        "site",
        "model",
        "defect_category",
        "defect_qty",
    ]
].copy()
df_work
```

```python
#cell5
df_work_positive = df_work[
    df_work["defect_qty"] > 0
].copy()

df_work_positive
```

```python
df_category_unique = df_work_positive.drop_duplicates(
    [
        "site",
        "model",
        "defect_category",
    ]
)
df_category_unique
```

```python
#cell7
df_category = (
    df_category_unique.groupby(
        [
            "site",
            "model",
        ],
        as_index=False,
    )["defect_category"].agg("/".join)
)
df_category
```

```python
df_category = df_category.rename(
    columns={
        "defect_category": "defect_category_list",
    }
)
df_category
```

```python
#cell9
df_result = pd.merge(
    df_plan,
    df_category,
    on=[
        "site",
        "model",
    ],
    how="left",
)
df_result
```

```python
df_result["defect_category_list"] = (
    df_result["defect_category_list"].fillna("発生なし")
)
df_result
```
