```python
df_production = df_quality.iloc[:, 2::2]
df_production
```

```python
df_defects = df_quality.iloc[:, 3::2]
df_defects
```

```python
df_info = df_quality.iloc[:,0:2]
df_info
```

```python
df_ppm = df_info.copy()
df_ppm
```

```python
df_ppm["melmb_ppm"] = (
    df_quality["melmb_defects"] / df_quality["melmb_production"] * 1000000
).round(1)
df_ppm
```

```python
df_ppm["wse_ppm"] = (
    df_quality["wse_defects"] / df_quality["wse_production"] * 1000000
).round(1)
df_ppm
```

```python
df_ppm["gse_ppm"] = (
    df_quality["gse_defects"] / df_quality["gse_production"] * 1000000
).round(1)
df_ppm
```

```python
df_ppm_only = df_ppm.iloc[:, 2:]
df_ppm_only
```

```python
df_ppm["max_ppm"] = df_ppm_only.max(axis=1)
df_ppm
```

```python
df_ppm["worst_site"] = df_ppm_only.idxmax(axis=1)
df_ppm
```

```python
df_ppm["危険ランク"] = np.where(
    df_ppm["max_ppm"] >= 2500,
    "s級",
    np.where(
        df_ppm["max_ppm"] >= 1000,
        "a級",
        "通常"
    )
)
df_ppm
```

```python
df_ppm_sorted = df_ppm.sort_values("max_ppm", ascending=False)
df_ppm_sorted
```

```python
df_danger = df_ppm_sorted[df_ppm_sorted["危険ランク"] != "通常"]
df_danger
```

```python
df_ppm_sorted.to_csv(
    "output/ppm_quality_iloc_ppm_all.csv",
    index=False,
    encoding="utf-8-sig"
)

df_danger.to_csv(
    "output/quality_iloc_ppm_danger.csv",
    index=False,
    encoding="utf=8-sig"
)
```
