## 演習問題２

```python
import pandas as pd

df_quality = pd.DataFrame({
    "site": ["melmb", "wse", "gse", "ise"],
    "production": [10000, 12000, 9000, 8000],
    "defects": [5, 1, 9, 2]
})

df_quality["ppm"] = (df_quality["defects"] / df_quality["production"] * 1000000).round(1)

df_quality
```
