## 実務演習

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
     '拠点': ['MELMB', 'WSE', 'GSE', 'ISE', 'SUS', 'MELMB', 'WSE', 'GSE', 'ISE', 'SUS'],
    '製品': ['LCM', 'LCM', 'LCM', 'LCM', 'LCM', 'H/L', 'H/L', 'H/L', 'H/L', 'H/L'],
    '不具合名': ['ネジゆるみ', 'ハンダ不足', 'コネクタずれ', 'ラベル間違い', '不点灯',
              '異音', 'キズ', '水入り', '点灯ちらつき', '通信エラー'],
    '生産数': [10000, 12000, 9000, 8000, 15000, 7000, 11000, 9500, 6000, 13000],
    '不良数': [5, 12, 9, 2, 20, 1, 15, 8, 10, 6],
    '重要度': ['中', '高', '高', '低', '高', '低', '中', '高', '高', '中']
})
```

```python
df.head()
```

```python
df.tail()
```

```python
df["拠点"]
```

```python
df["拠点"].unique()
```

```python
df["ppm"] = df["不良数"] / df["生産数"] * 1000000
df["ppm"] = df["ppm"].round(1)
df.head()
```

```python
df_wse = df[df["拠点"] == "WSE"]
df_wse
```

```python
defect_ranking = df_wse.loc[:, ["拠点", "製品", "不具合名", "不良数", "ppm", "重要度"]]
defect_ranking
```

```python
df["危険度"] = np.where(df["ppm"] >= 1000, "危険モンスター", "通常モンスター")
df.sort_values(by="ppm", ascending=False).head(10)
```
