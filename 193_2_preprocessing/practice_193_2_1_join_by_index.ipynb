## 実務練習問題

```python
import pandas as pd
import numpy as np
import os

os.makedirs("input", exist_ok=True)
```

```python
quality_data = pd.DataFrame({
     'date': ['2026/6/1', '2026/6/1', '2026/6/1', '2026/6/2', '2026/6/2', '2026/6/2'],
    'site': ['MELMB', 'WSE', 'GSE', 'MELMB', 'WSE', 'ISE'],
    'defect': ['点灯不良', '外観不良', 'ショート', '点灯不良', '外観不良', 'コネクタ浮き'],
    'production': [10000, 8000, 12000, 11000, 8500, 9000],
    'defects': [3, 5, 1, 4, 6, 2]
})

site_master = pd.DataFrame({
      'site': ['MELMB', 'WSE', 'GSE', 'ISE'],
    'country': ['Thailand', 'China', 'China', 'Japan'],
    'region': ['Asia', 'Asia', 'Asia', 'Japan'],
    'owner': ['Lighting Team', 'Appearance Team', 'Circuit Team', 'Connector Team']
})
```

```python
quality_data.to_csv("input/quality_monsters.csv", index=False, encoding="Shift_JIS")
site_master.to_csv("input/site_master.csv", index=False, encoding="Shift_JIS")
```

```python
df = pd.read_csv("input/quality_monsters.csv", encoding="Shift_JIS")
df.head()
```

```python
df_sites = pd.read_csv("input/site_master.csv",
                      encoding="Shift_JIS",
                      index_col=0)
df_sites.head()
```

```python
df_join = df.join(df_sites, on="site")
df_join.head()
```

```python
df_join["ppm"] = (df_join["defects"] / df_join["production"] * 1000000).round(1)
```

```python
df_join.head()
```

```python
df_join["判定"] = np.where(df_join["ppm"] >= 500, "要確認", "通常")
```

```python
df_join.head()
```

```python
df_ranking = df_join.sort_values(by="ppm", ascending=False)
```

```python
df_ranking
```

## 「193-2: 複数のDataFrameを結合しよう」 2問目

```python
import pandas as pd

df = pd.read_csv("input/data.csv", encoding="Shift_JIS")
```

```python
df_search1 = df[df["temp"] >= 18]
df_search1.head()
```

```python
df_search2 = df[(df["temp"] > 18) & (df["temp"] < 20)]
df_search2.head()
```

```python
df_search3 = df[(df["station"].isin(["東京", "札幌"])) & (df["temp"] < 20)]
df_search3.head()
```
