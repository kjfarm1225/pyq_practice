```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        ["2026-06-01", "08:00", "J69P", "MELMB", "Line1", "L250601-A", 300, 0],
        ["2026-06-01", "10:00", np.nan, np.nan, np.nan, np.nan, 280, 1],
        ["2026-06-01", "13:00", np.nan, np.nan, np.nan, np.nan, 320, 2],
        ["2026-06-02", "08:00", "310D", "WSE", "Line2", "L250602-B", 250, 0],
        ["2026-06-02", "10:00", np.nan, np.nan, np.nan, np.nan, 260, 1],
        ["2026-06-02", "13:00", np.nan, np.nan, np.nan, np.nan, 240, 0],
        ["2026-06-03", "08:00", "J69P", "GSE", "Line1", "L250603-C", 310, 3],
        ["2026-06-03", "10:00", np.nan, np.nan, np.nan, np.nan, 290, 0],
        ["2026-06-03", "13:00", np.nan, np.nan, np.nan, np.nan, 300, 1],
    ],
    columns=[
        "inspection_date",
        "inspection_time",
        "project",
        "site",
        "line",
        "lot_no",
        "inspection_qty",
        "defect_qty",
    ],
)
df
```

```python
df_work = df.copy()
df_work
```

```python
df_work["project_filled"] = df_work["project"].ffill()
df_work
df_work["site_filled"] = df_work["site"].ffill()
df_work
df_work["line_filled"] = df_work["line"].ffill()
df_work
df_work["lot_no_filled"] = df_work["lot_no"].ffill()
df_work
```

```python
df_result = df_work[
    [
        "inspection_date",
        "inspection_time",
        "project_filled",
        "site_filled",
        "line_filled",
        "lot_no_filled",
        "inspection_qty",
        "defect_qty",
    ]
].copy()
df_result
```

```python
df_result["defect_rate"] = df_result["defect_qty"] / df_result["inspection_qty"]
df_result
```

```python
df_result_sort = df_result.sort_values(
    "defect_rate",
    ascending=False,
)
df_result_sort
```
