#### 演習問題

```python
import pandas as pd
import numpy as np
import os

os.makedirs("input", exist_ok=True)
os.makedirs("output", exist_ok=True)

df = pd.DataFrame(
    [
        ["2026/06/01", "MELMB", "LCM-A", "点灯不良", 10000, 3],
        ["2026/06/01", "WSE", "LCM-B", "外観不良", 8000, 5],
        ["2026/06/02", "GSE", "LCM-C", "ショート", 12000, 1],
        ["2026/06/02", "MELMB", "LCM-A", "点灯不良", 11000, 8],
        ["2026/06/03", "ISE", "LCM-D", "コネクタ浮き", 9000, 2],
        ["2026/06/03", "WSE", "LCM-B", "異物混入", 7000, 7],
    ],
    columns=["date", "site", "product", "defect", "production", "defects"],
)
```

```python
df.to_csv("input/quality_monsters.csv", index=False, encoding="Shift_JIS")
print(df)
```

```python
df_loaded = pd.read_csv("input/quality_monsters.csv", encoding="Shift_JIS")
print(df_loaded)

df_loaded["ppm"] = (df_loaded["defects"] / df_loaded["production"] * 1000000).round(2)
print(df_loaded)

df_loaded["判定"] = np.where(
    df_loaded["ppm"] >= 1000,
    "危険モンスター",
    "通常もんすたー"
)

df_sorted = df_loaded.sort_values(by="ppm", ascending=False)
print(df_sorted)

df_report = df_sorted[["date", "site", "product", "defect", "production", "defects", "ppm", "判定"]]
print(df_report)
df_report.to_csv("output/quality_monster_report.csv", index=False, encoding="utf-8-sig")

df_danger = df_report[df_report["判定"] == "危険モンスター"]
print(df_danger)

df_danger.to_csv("output/dangerous_monsters.csv", index=False, encoding="utf-8-sig")
print(df_report)
```
