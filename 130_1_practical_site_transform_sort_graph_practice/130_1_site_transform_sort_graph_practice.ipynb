```python
#cell1
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
plt.rcParams["font.family"] = "Meiryo"
```

```python
#cell2
df = pd.DataFrame(
    [
        ["SUS", "J69P", "コイル", 12000, 3],
        ["SUS", "J69P", "Buck-IC", 10000, 1],
        ["WSE", "310D", "コイル", 15000, 2],
        ["WSE", "310D", "NTF", 14000, 0],
        ["GSE", "J69P", "コイル", 9000, 4],
        ["GSE", "J69P", "Buck-IC", 8000, 2],
        ["MELMB", "310D", "コイル", 20000, 5],
        ["MELMB", "310D", "NTF", 18000, 1],
        ["ISE", "J69P", "コイル", 11000, 3],
        ["ISE", "J69P", "NTF", 9000, 0],
    ],
    columns=[
        "site",
        "model",
        "defect_category",
        "inspection_qty",
        "defect_qty",
    ],
)

df
```

```python
df_work = df.copy()
df_work
```

```python
df_work["site_total_defect_qty"] = (
    df_work.groupby("site")["defect_qty"].transform("sum")
)
df_work
```

```python
#cell5 
df_work["site_total_inspection_qty"] = (
    df_work.groupby("site")["inspection_qty"].transform("sum")
)
df_work
```

```python
df_work["site_defect_rate"] = (
    df_work["site_total_defect_qty"]
    / df_work["site_total_inspection_qty"]
)
df_work
```

```python
df_work["site_defect_ppm"] = (
    df_work["site_defect_rate"] * 1000000
)
df_work
```

```python
#cell 8
df_site_check = df_work[
    [
         "site",
        "site_total_defect_qty",
        "site_total_inspection_qty",
        "site_defect_rate",
        "site_defect_ppm",
    ]
].drop_duplicates()
df_site_check
```

```python
df_site_report = df_site_check.sort_values(
    "site_defect_ppm",
    ascending=False,
).reset_index(
    drop=True
)
df_site_report
```

```python
#cell10

plt.figure(
    figsize=(10, 5)
)

plt.bar(
    df_site_report["site"],
    df_site_report["site_defect_ppm"],
)

plt.title("site別defect_ppm")
plt.xlabel("site")
plt.ylabel("site_defect_ppm")
plt.show()
```
