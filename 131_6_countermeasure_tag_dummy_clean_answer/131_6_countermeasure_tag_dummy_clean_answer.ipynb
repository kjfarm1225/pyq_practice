```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "，",
    ",",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    " ",
    "",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "　",
    "",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "外観チェック",
    "外観検査",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "通電確認",
    "通電検査",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "教育実施",
    "教育",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "設変",
    "設計変更",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df["countermeasure_tag_clean"] = df["countermeasure_tag_clean"].str.replace(
    "作業標準改定",
    "作業標準改訂",
    regex=False,
)
df[["countermeasure_tag", "countermeasure_tag_clean"]]
```

```python
df[
    [
        "defect_id",
        "countermeasure_tag",
        "countermeasure_tag_clean",
    ]
]
```

```python
df_measure_dummy = df["countermeasure_tag_clean"].str.get_dummies(sep=",")
df_measure_dummy
```

```python
df = df.join(df_measure_dummy)
```

```python
df
```

```python
measure_cols = df_measure_dummy.columns.tolist()
```

```python
measure_cols
```

```python
df_measure_count = df[measure_cols].sum().reset_index()
df_measure_count
```

```python
df_measure_count.columns = ["measure", "count"]
```

```python
df_measure_count
```

```python
df_measure_count = df_measure_count.sort_values(
    "count",
    ascending=False,
).reset_index(drop=True)
df_measure_count
```

```python
plt.figure(figsize=(8, 4))
```

```python
plt.bar(
    df_measure_count["measure"],
    df_measure_count["count"],
)

plt.title("対策タグ別件数")
plt.xlabel("対策タグ")
plt.ylabel("件数")
plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

```python
df_site_visual_count = df.groupby(
    "site",
    as_index=False,
)["外観検査"].sum()
df_site_visual_count
```

```python
df_site_visual_count = df_site_visual_count.rename(
    columns={
        "外観検査": "visual_check_count",
    }
)
df_site_visual_count
```

```python
df_site_visual_count = df_site_visual_count.sort_values(
    ["visual_check_count","site"],
    ascending=[False,True]
)
df_site_visual_count
```

```python
df_month_design_count = df.groupby(
    "occ_month",
    as_index=False,
)["設計変更"].sum()
df_month_design_count
```

```python
df_month_design_count = df_month_design_count.rename(
    columns={
        "設計変更": "design_change_count",
    }
)
df_month_design_count
```

```python
plt.figure(figsize=(8, 4))
```

```python
plt.plot(
    df_month_design_count["occ_month"],
    df_month_design_count["design_change_count"]
)

plt.title("月別設計変更件数")
plt.xlabel("発生月")
plt.ylabel("設計変更件数")
plt.grid(True)

plt.tight_layout()
plt.show()
```
