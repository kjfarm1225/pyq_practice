```python
import pandas as pd
```

```python
df_ticket = pd.DataFrame(
    [
        ["T001", "J69P", "MELMB/Line1/組立", "コイル/鉄粉/ショート", "RH Lo不灯", 2],
        ["T002", "J69P", "MELMB/Line1/検査", "コイル/巻線ばらつき", "RH Hi/Lo不灯", 3],
        ["T003", "J69P", "MELMB/Line2/組立", "Buck-IC/焼損", "全灯火不点灯", 5],
        ["T004", "310D", "WSE/Line1/検査", "異物/混入", "点灯不安定", 1],
        ["T005", "310D", "WSE/Line2/組立", "はんだ/クラック", "点灯不安定", 4],
        ["T006", "J69P", "GSE/Line1/組立", "コイル/鉄粉/ショート", "RH Lo不灯", 2],
        ["T007", "310D", "GSE/Line2/検査", "NTF", "再現せず", 1],
        ["T008", "J69P", "ISE/Line1/組立", "コイル/巻線ばらつき", "RH Lo不灯", 3],
        ["T009", "310D", "ISE/Line1/検査", "異物/付着", "点灯不安定", 2],
        ["T010", "J69P", "WSE/Line2/検査", "Buck-IC/焼損", "RH Lo不灯", 4],
        ["T011", "310D", "MELMB/Line2/組立", "はんだ/濡れ不足", "点灯不安定", 3],
        ["T012", "J69P", "GSE/Line1/検査", "コイル/鉄粉/ショート", "RH Hi/Lo不灯", 5],
    ],
    columns=[
        "ticket_id",
        "project",
        "発生場所",
        "原因タグ",
        "symptom",
        "risk_score",
    ],
)
df_ticket.head(3)
```

```python
df_site_master = pd.DataFrame(
    [
         ["MELMB", "メルモビ", "品質保証1課", "重点監視"],
        ["WSE", "中国WSE", "品質保証2課", "重点監視"],
        ["GSE", "中国GSE", "品質保証2課", "通常監視"],
        ["ISE", "日本ISE", "品質保証1課", "通常監視"],
    ],
    columns=[
        "拠点",
        "拠点名",
        "担当部署",
        "監視区分",
    ],
)
df_site_master
```

```python
df_ticket[["拠点", "ライン", "工程"]] = df_ticket["発生場所"].str.split("/", expand=True)
df_ticket.head(3)
```

```python
df_ticket[["原因1", "原因2", "原因3"]] = df_ticket["原因タグ"].str.split("/", expand=True)
df_ticket.head(3)
```

```python
df_ticket[["原因1","原因2", "原因3"]] = df_ticket[["原因1","原因2", "原因3"]].fillna("")
df_ticket
```

```python
df_cross = df_ticket.pivot_table(
    values="ticket_id",
    index="拠点",
    columns="原因1",
    aggfunc="count",
    fill_value=0,
)
df_cross
```

```python
df_cross = df_cross[
    [
        "Buck-IC",
        "NTF",
        "はんだ",
        "コイル",
        "異物"
    ]
].copy()
```

```python
df_cross["total"] = df_cross.sum(axis=1)
df_cross
```

```python
df_cross_sort = df_cross.sort_values(
    "total",
    ascending=False,
)
df_cross_sort
```

```python
df_cross_sort = df_cross_sort.reset_index()
df_cross_sort
```

```python
df_report = df_cross_sort.merge(
    df_site_master,
    on="拠点"
)
df_report
```

```python
df_report = df_report[
    [
        "担当部署",
        "拠点名",
        "拠点",
        "監視区分",
        "Buck-IC",
        "NTF",
        "はんだ",
        "コイル",
        "異物",
        "total",
    ]
].copy()
df_report.head(3)
```

```python
df_report["check_level"] = "通常確認"
#df_report.head(1)

df_report.loc[
    df_report["total"] >= 4,
    "check_level",
] = "全体件数重点確認"
#df_report

df_report.loc[
    df_report["コイル"] >= 2,
    "check_level",
] = "コイル重点確認"
df_report

df_report.loc[
    df_report["Buck-IC"] >= 1,
    "check_level",
] = "Buck-IC確認"
df_report
```

```python
df_priority = df_report[
    df_report["check_level"] != "通常確認"
].copy()

df_priority
```
