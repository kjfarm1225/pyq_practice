```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

df_log = pd.read_csv("dataset/product_log.csv")
df_log.head()
```

```python
df_log.plot(marker="*")
```

```python
df_log["Num_fill_0"] = df_log["Num"].fillna(0)
df_log[16:24]
```

```python
df_log["Num"].plot(marker="*")
df_log["Num_fill_0"].plot()
plt.legend()
```

```python
df_log.fillna(0)
print("欠損値の数 =", df_log["Num"].isnull().sum())
```

```python
df_log_copy = df_log.copy()
df_log_copy.fillna(0, inplace=True)
print("欠損値の数 =", df_log_copy["Num"].isnull().sum())
```

```python
df_log_copy["Num"].plot(marker="*")
df_log_copy["Num_fill_0"].plot()
plt.legend()
```

```python
df_log["Num_fill_minus1"] = df_log["Num"].fillna(-1)
df_log["Num_fill_minus1"]
```

```python
df_log["Product_fill_unknown"] = df_log["Product"].fillna("不明")
df_log["Product_fill_unknown"]
```
