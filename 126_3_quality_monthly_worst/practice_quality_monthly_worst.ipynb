```python
import pandas as pd
import numpy as np
from pathlib import Path
import os

Path("output").mkdir(exist_ok=True)
```

```python
df_quality = pd.DataFrame(
    [
        ["2026-05-01", "MELMB", "Lamp-A", "点灯不良", 12000, 18],
        ["2026-05-03", "MELMB", "Lamp-B", "外観キズ", 8500, 4],
        ["2026-06-01", "MELMB", "Lamp-A", "点灯不良", 11000, 30],
        ["2026-06-04", "MELMB", "Lamp-C", "寸法NG", 9000, 8],

        ["2026-05-02", "WSE", "Lamp-D", "異物混入", 10000, 25],
        ["2026-05-05", "WSE", "Lamp-E", "はんだ不良", 7000, 5],
        ["2026-06-02", "WSE", "Lamp-D", "異物混入", 9500, 10],
        ["2026-06-06", "WSE", "Lamp-F", "点灯不良", 6000, 18],

        ["2026-05-04", "GSE", "Lamp-G", "外観キズ", 8000, 2],
        ["2026-05-08", "GSE", "Lamp-H", "寸法NG", 5000, 11],
        ["2026-06-03", "GSE", "Lamp-G", "外観キズ", 9000, np.nan],
        ["2026-06-07", "GSE", "Lamp-H", "寸法NG", 5200, 3],
    ],
    columns=["date", "site", "product", "defect_type", "production", "defects"],
)

print(df_quality)
df_quality["date"].dtype
```

```python
#date列を日付型に変換
df_quality["date"] = pd.to_datetime(df_quality["date"])
print(df_quality)
df_quality["date"].dtype
```

```python
#month列を作成
df_quality["month"] = df_quality["date"].dt.strftime("%Y-%m")
print(df_quality)
```

```python
df_quality["ppm"] = df_quality["defects"] / df_quality["production"] * 1000000
df_quality["ppm"] = df_quality["ppm"].round(1)

print(df_quality)
```

```python
#6危険ランクを付ける
df_quality["危険ランク"] = np.where(
    df_quality["ppm"] >= 2000,
    "s級",
    np.where(
        df_quality["ppm"] >= 1000,
        "a級",
        "通常"
    )
)
df_quality
```

```python
#7：site・month・PPMで並び替え
df_sorted = df_quality.sort_values(
    ["site", "month", "ppm"],
    ascending=[True, True, False]
)

df_sorted
```

```python
#8：危険データだけ抽出
df_danger = df_sorted[df_sorted["危険ランク"] != "通常"]
df_danger
```

```python
#9：CSV保存
df_sorted.to_csv(
    "output/quality_monthly_worst_all.csv",
    index=False,
    encoding="utf-8-sig"
)

df_danger.to_csv(
    "output/quality_monthly_worst_danger.csv",
    index=False,
    encoding="utf-8-sig"
)
```
