```python
import numpy as np
import pandas as pd
```

```python
rnd = np.random.default_rng(3)
df = pd.DataFrame(
    rnd.standard_normal(100),
    columns=[
        "Value",
    ],
)
df
```

```python
bins = [
    -np.inf,
    -2,
    -1,
    0,
    1,
    2,
    np.inf,
]
```

```python
df["Category"] = pd.cut(
    df["Value"],
    bins,
)
df
```
