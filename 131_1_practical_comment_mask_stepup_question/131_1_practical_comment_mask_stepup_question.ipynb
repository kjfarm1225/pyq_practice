```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.DataFrame(
    [
        ["D-001", "SUS", "J69P", "2026/06/15にKLAK-67890で3件発生。160V条件で確認。"],
        ["D-002", "WSE", "310D", "2026/06/18にKLBC-12345で1件発生。140V検査では検出せず。"],
        ["D-003", "GSE", "J69P", "2026/06/20にPY-110相当品で2件再現。12V条件で確認。"],
        ["D-004", "MELMB", "310D", "現象再現せず。対象品番KLEE-55555は継続確認中。"],
        ["D-005", "SUS", "310D", "2026/06/25にLCM-220で5件確認。160V条件で全数検査中。"],
    ],
    columns=[
        "defect_id",
        "site",
        "model",
        "comment",
    ],
)
df_defect
```

```python
df_work = df_defect.copy()
df_work
```

```python
df_work["comment_public"] = df_work["comment"]
```

```python
df_work["comment_public"] = df_work["comment_public"].str.replace(
    r"\d{4}/\d{2}/\d{2}",
    "[DATE]",
    regex=True
)
df_work
```

```python
df_work["comment_public"] = df_work["comment_public"].str.replace(
    r"[A-Z]+-\d+",
    "[PART]",
    regex=True
)
df_work
```

```python
df_work["comment_public"] = df_work["comment_public"].str.replace(
    r"\d+件",
    "[QTY]",
    regex=True,
)
```

```python
df_work
```

```python
df_work["comment_public"] = df_work["comment_public"].str.replace(
    r"\d+V",
    "[VOLT]",
    regex=True,
)
```

```python
df_work
```

```python
df_result = df_work[
    [
        "defect_id",
        "site",
        "model",
        "comment",
        "comment_public",
    ]
]
df_result
```
