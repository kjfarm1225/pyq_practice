```python
import pandas as pd
df = pd.read_table("dataset/review_after.tsv", encoding="shift_jis")
grouped = df.groupby(["シリーズ名", "バージョン番号"])
agg_df = grouped["評価"].agg(["mean", "size"])
agg_df = agg_df.rename(columns={"mean":"評価の平均", "size":"レビュー件数"})
agg_df["全体順位"] = agg_df["評価の平均"].rank(ascending=False)
agg_df
```
