```python
import numpy as np

prices = np.array([[100, 130, 180],
                   [110, 135, 190]])

quantities = np.array([20, 10, 5])

print(prices.shape)
print(quantities.shape)
```

```python
total_prices = prices @ quantities
```

```python
total_prices
```

```python
total_prices = np.dot(prices, quantities)
total_prices
```
