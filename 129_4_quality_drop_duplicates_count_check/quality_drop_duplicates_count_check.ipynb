```python
#cell3
df_work = df.copy()
df_work
```

```python
#cell4
df_work["update_time"] = pd.to_datetime(
    df_work["update_time"]
)
df_work
```

```python
#cell5
df_duplicate_check = df_work[
    df_work.duplicated(
    "defect_id",
    keep=False,
    )
].copy()
df_duplicate_check
```

```python
#call6
before_count = len(df_work)
before_count
```

```python
#cell7
df_sort = df_work.sort_values(
    "update_time"
)
df_sort
```

```python
df_deleted = df_sort[
    df_sort.duplicated(
        "defect_id",
        keep="last",
    )
].copy()
df_deleted
```

```python
#cell9
df_result = df_sort.drop_duplicates(
    "defect_id",
    keep="last",
)
df_result
```

```python
after_count = len(df_result)
after_count
```

```python
deleted_count = before_count - after_count
deleted_count
```

```python
#cell12

duplicated_id_count = df_duplicate_check["defect_id"].nunique()

df_count_summary = pd.DataFrame(
    [
        ["削除前行数", before_count],
        ["削除後行数", after_count],
        ["削除行数", deleted_count],
        ["重複していたdefect_id数", duplicated_id_count], 
    ],
    columns=[
        "確認項目",
        "件数",
    ],
)
df_count_summary
```

```python
#cell13
df_site_total = df_result.pivot_table(
    values="defect_qty",
    index="site",
    aggfunc="sum",
)
df_site_total
```

```python
#cell14
df_site_total_sort = df_site_total.sort_values(
    "defect_qty",
    ascending=False,
)
df_site_total_sort
```
