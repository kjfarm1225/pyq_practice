```python
np.mod(np.arange(df.shape[0] * 2), 2)
```

```python
L = np.c_[
    np.c_[df,df].reshape(-1, df.shape[1]),
    np.mod(np.arange(df.shape[0] * 2), 2)
]
L
```

```python
L.T[0]
```

```python
L.T[-1]
```

```python
L.T[0] ^ L.T[-1]
```

```python
L = np.c_[L, L.T[0] ^ L.T[-1]]
L
```

```python
df = pd.DataFrame(L)
df
```
