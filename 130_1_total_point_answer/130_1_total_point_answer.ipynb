```python
#cell1

import pandas as pd
```

```python
#cell2 
df = pd.DataFrame(
    [
        ["Alice", "国語", 87],
        ["Alice", "数学", 72],
        ["Bob", "国語", 65],
        ["Bob", "数学", 92],
    ],
    columns=[
        "Name",
        "Subject",
        "Point",
    ],
)
df
```

```python
#cell3
sr = df.groupby("Name")["Point"].sum()
sr
```

```python
#cell4 
df_result = df.join(
    sr.rename("Total"),
    on="Name",
)
df_result
```
