```python
#cell1
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
#cell2
df_site = pd.DataFrame(
    [
        ["SUS", "North America", "北米拠点"],
        ["WSE", "China", "中国拠点"],
        ["GSE", "China", "中国拠点"],
        ["MELMB", "China", "中国拠点"],
        ["ISE", "Japan", "日本拠点"],
    ],
    columns=[
        "site",
        "area",
        "site_group",
    ],
)
df_site
```

```python
df_defect = pd.DataFrame(
    [
        ["SUS", "J69P", "コイル", 3],
        ["SUS", "J69P", "Buck-IC", 1],
        ["SUS", "J69P", "コイル", 2],
        ["SUS", "310D", "NTF", 1],
        ["WSE", "310D", "コイル", 2],
        ["WSE", "310D", "NTF", 0],
        ["GSE", "J69P", "コイル", 4],
        ["GSE", "J69P", "Buck-IC", 2],
        ["MELMB", "310D", "コイル", 5],
        ["MELMB", "310D", "NTF", 1],
    ],
    columns=[
        "site",
        "model",
        "defect_category",
        "defect_qty",
    ],
)

df_defect
```

```python
#cell4

df_category = (
    df_defect.groupby(
        "site",
        as_index=False,
    )["defect_category"]
    .agg(lambda x: "/".join(x.drop_duplicates()))
    .rename(
        columns={
            "defect_category": "defect_category_list",
        }
    )
)

df_category
```

```python
#cell5 
df_result = pd.merge(
    df_site,
    df_category,
    on="site",
    how="left",
)
df_result
```

```python
#cell6
df_result["defect_category_list"] = (
    df_result["defect_category_list"].fillna("発生無し")
)
df_result
```

```python
#cell4-1
df_category_base = df_defect[
    [
        "site",
        "defect_category",
    ]
].copy()
df_category_base
```

```python
#cell 4-2
df_category_unique = df_category_base.drop_duplicates()
df_category_unique
```

```python
#4-3
df_category = (
    df_category_unique.groupby(
        "site",
        as_index=False,
    )["defect_category"].agg("/".join)
)
df_category
```

```python
#4-4
df_category =df_category.rename(
    columns={
        "defect_category": "defect_category_list",
    }
)
df_category
```
