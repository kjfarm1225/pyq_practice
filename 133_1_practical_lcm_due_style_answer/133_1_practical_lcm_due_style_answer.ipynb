```python
import pandas as pd
```

```python
df = pd.read_csv(
    "dataset/lcm_due_style_practice.csv",
)
df
```

```python
df["defect_count_num"] = pd.to_numeric(
    df["defect_count"],
    errors="coerce",
)
df
```

```python
df["due_date_raw"] = df["due_date"]
df.dtypes
```

```python
df["due_date_clean"] = (
    df["due_date"]
    .astype(str)
    .str.strip()
    .str.replace("/","-", regex=False)
)

df["due_date_dt"] = pd.to_datetime(
    df["due_date_clean"],
    errors="coerce",
)
df.dtypes
```

```python
base_date = pd.Timestamp("2026-08-05")
```

```python
df["defect_count_num"]
```

```python
df_defect_count_ng = df[
    df["defect_count_num"].isna()
].copy()
display(df_defect_count_ng)
```

```python
df["due_date_dt"]
```

```python
df_due_date_ng = df[
    df["due_date_dt"].isna()
].copy()
display(df_due_date_ng)
```

```python
df_work = df[
    df["defect_count_num"].notna()
    & df["due_date_dt"].notna()
].copy()
df_work
```

```python
display_cols = [
    "no",
    "site",
    "model",
    "defect_category",
    "defect_count",
    "due_date",
    "status",
    "priority",
    "comment",
]

df_display = df[display_cols].copy()

df_display
```

```python
#cell 9
def style_defect_count(x):
    value = pd.to_numeric(
        x,
        errors="coerce",
    )

    if pd.isna(value):
        style = "background-color: lightgray; color: black;"
    elif value >= 5:
        style = "font-weight: bold;"
    else:
        style = None

    return style
```

```python
#cell 10
def style_due_date(x):
    value_clean = (
        str(x)
        .strip()
        .replace("/", "-")
    )

    value = pd.to_datetime(
        value_clean,
        errors="coerce",
    )

    if pd.isna(value):
        style = "background-color: lightgray; color: black;"
    elif value < base_date:
        style = "color: red; font-weight: bold;"
    else:
        style = None

    return style
```

```python
#cell 11
def style_status(x):
    if x == "未対応":
        style = "background-color: yellow; color: black;"
    elif x == "不明":
        style = "background-color: lightgray; color: black;"
    else:
        style = None

    return style
```

```python
#cell 12
def style_priority(x):
    if x == "高":
        style = "font-weight: bold;"
    else:
        style = None

    return style
```

```python
styled_df = (
    df_display.style
    .map(style_defect_count, subset=["defect_count"])
    .map(style_due_date, subset=["due_date"])
    .map(style_status, subset=["status"])
    .map(style_priority, subset=["priority"])
)

display(styled_df)
```
