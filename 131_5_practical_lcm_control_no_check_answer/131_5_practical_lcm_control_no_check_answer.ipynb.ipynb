```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.read_csv("dataset/lcm_control_no_check.csv")
```

```python
pattern = r"[A-Z]{3}-\d{6}-\d{3}"
```

```python
df_defect["control_no_ok"] = df_defect["control_no"].str.fullmatch(
    pattern,
    na=False,
)
```

```python
df_valid = df_defect[
    df_defect["control_no_ok"]
].copy()
df_valid
```

```python
df_invalid = df_defect[
    ~df_defect["control_no_ok"]
].copy()
df_invalid
```

```python
total_count = len(df_defect)
print(total_count)

valid_count = len(df_valid)
print(valid_count)

invalid_count = len(df_invalid)
print(invalid_count)

valid_rate = round(
    valid_count / total_count * 100,
    1,
)
print(valid_rate)
```

```python
df_check_summary = pd.DataFrame(
    {
            "item": [
                "全データ件数",
                "書式ok件数",
                "書式ng件数",
                "書式ok率",
            ],
            "value": [
                total_count,
                valid_count,
                invalid_count,
            valid_rate,
            ],
    }
)
df_check_summary
```

```python
df_site_valid_qty = df_valid.groupby(
    "site",
    as_index=False,
)["defect_qty"].sum()
df_site_valid_qty
```

```python
df_site_valid_qty = df_site_valid_qty.rename(
    columns={
        "defect_qty": "valid_defect_qty",
    }
)
df_site_valid_qty
```

```python
df_site_valid_qty = df_site_valid_qty.sort_values(
    ["valid_defect_qty", "site"],
    ascending=[False, True],
).reset_index(drop=True)
df_site_valid_qty
```

```python
df_invalid_check = df_invalid[
    [
        "defect_id",
        "site",
        "occ_month",
        "control_no",
        "comment",
    ]
].copy()
df_invalid_check
```
