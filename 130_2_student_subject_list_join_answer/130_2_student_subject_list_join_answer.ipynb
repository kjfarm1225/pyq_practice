```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
df_class = pd.DataFrame(
    [
        ["Alice", "1-A"],
        ["Bob", "1-B"],
    ],
    columns=["Name", "Class"],
)
df_class
```

```python
df = pd.DataFrame(
    [
        ["Alice", "国語", 87],
        ["Alice", "数学", 72],
        ["Bob", "国語", 65],
        ["Bob", "理科", 92],
    ],
    columns=[
        "Name",
        "Subject",
        "Point",
    ],
)
df
```

```python
sr = df.groupby("Name").Subject.apply("/".join)
sr
```

```python
df_result = df_class.join(
    sr,
    "Name",
)

df_result
```
