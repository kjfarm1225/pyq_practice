## 実務練習問題

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("input/quality_data.csv", encoding="Shift_JIS")
```

```python
series_defects = df["defects"]
series_defects
```

```python
mean_defects = series_defects.mean()
mean_defects
```

```python
series_filled = series_defects.fillna(mean_defects)
series_filled
```

```python
df["defects"] = series_filled
```

```python
df["defects"]
```

```python
df["ppm"] = (df["defects"] / df["production"] * 1000000).round(1)
```

```python
df["危険度"] = np.where(df["ppm"] >= 500, "要確認", "通常")

df_ranking = df.sort_values(by="ppm", ascending=False)
df_ranking
```
