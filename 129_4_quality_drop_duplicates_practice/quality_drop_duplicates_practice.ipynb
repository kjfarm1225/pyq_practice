```python
#cell1 
import pandas as pd
```

```python
df = pd.DataFrame(
    [
         ["D001", "J69P", "MELMB", "コイル", 1, "2026-06-01 09:00"],
        ["D002", "J69P", "WSE", "Buck-IC", 0, "2026-06-01 09:10"],
        ["D003", "310D", "GSE", "はんだ", 1, "2026-06-01 09:20"],
        ["D004", "J69P", "MELMB", "コイル", 3, "2026-06-01 09:30"],
        ["D002", "J69P", "WSE", "Buck-IC", 2, "2026-06-01 10:00"],
        ["D005", "310D", "WSE", "NTF", 0, "2026-06-01 10:10"],
        ["D006", "J69P", "GSE", "コイル", 2, "2026-06-01 10:20"],
        ["D004", "J69P", "MELMB", "コイル", 4, "2026-06-01 10:30"],
        ["D007", "310D", "MELMB", "はんだ", 1, "2026-06-01 10:40"],
    ],
    columns=[
        "defect_id",
        "project",
        "site",
        "defect_category",
        "defect_qty",
        "update_time",
    ],
)

df
```

```python
df_work = df.copy()
df_work
```

```python
df_work["update_time"] = pd.to_datetime(
    df_work["update_time"]
)

df_work
```

```python
df_duplicate_check = df_work[
    df_work.duplicated(
        "defect_id",
        keep=False,
    )
].copy()
df_duplicate_check
```

```python
df_sort = df_work.sort_values(
    "update_time"
)
df_sort
```

```python
df_result = df_sort.drop_duplicates(
    "defect_id",
    keep="last",
)
df_result
```

```python
df_site_total = df_result.pivot_table(
    values="defect_qty",
    index="site",
    aggfunc="sum",
)
df_site_total
```

```python
df_site_total_sort = df_site_total.sort_values(
    "defect_qty",
    ascending=False,
)

df_site_total_sort
```
