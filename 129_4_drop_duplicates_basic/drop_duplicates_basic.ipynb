###### 129_4_drop_duplicates_basic

```python
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        ["Alice", "国語", 87],
        ["Alice", "数学", 0],
        ["Bob", "国語", 65],
        ["Bob", "数学", 92],
        ["Alice", "数学", 72],
    ],
    columns=["Name", "Subject", "Point"],
)
df
```

```python
df_result = df.drop_duplicates(
    ["Name", "Subject"],
    keep="last",
)
df_result
```
