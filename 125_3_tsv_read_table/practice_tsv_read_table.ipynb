###### TSVから品質危険ランキングを作ろう

```python
import pandas as pd
import numpy as np
import os

os.makedirs("output", exist_ok=True)
df = pd.read_table("input/quality_check_practice.tsv")
```

```python
df["ppm"] = df["defects"] / df["production"] * 1000000
df["ppm"] = df["ppm"].round(1)
```

```python
df["判定"]  = np.where(
    df["ppm"] >= 800,
    "危険モンスター",
    "通常モンスター"
)
```

```python
df_sorted = df.sort_values("ppm", ascending=False)
```

```python
df_danger = df_sorted[df_sorted["判定"] == "危険モンスター"]
```

```python
print(df_sorted)
df_sorted.to_csv("output/quality_report.csv", index=False, encoding="utf-8-sig")
print(df_danger)
df_danger.to_csv("output/danger_report.csv", index=False, encoding="utf-8-sig")

df_sorted
```
