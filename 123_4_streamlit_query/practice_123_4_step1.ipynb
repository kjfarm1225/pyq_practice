```python
import streamlit as st
import pandas as pd

# タイトル表示
st.header("数学の試験の成績の検索")

# CSVファイルを読み込む
df = pd.read_csv("input/exam.csv", encoding="utf-8")

# ユーザーにクエリ文字列を入力してもらう
s = st.text_input("クエリ")

# 入力があるときだけ実行
if s:
    try:
        # 入力されたクエリでデータを絞り込む
        result = df.query(s)

        # 結果を表で表示する
        st.dataframe(result)
    except:
        # クエリが間違っているときの表示
        st.error("クエリを正しくしてください")
        
```
