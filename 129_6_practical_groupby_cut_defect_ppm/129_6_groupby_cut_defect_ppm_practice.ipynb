```python
df_work = df.copy()
df_work
```

```python
df_summary = df_work.groupby(
    [
        "site",
        "defect_category",
    ],
    as_index=False,
)[
    [
        "defect_qty",
        "inspection_qty",
    ]
].sum()
df_summary
```

```python
df_summary = df_summary.rename(
    columns={
        "defect_qty": "total_defect_qty",
        "inspection_qty": "total_inspeciton_qty",
    }
)
df_summary
```

```python
#cell6
df_summary["defect_rate"] = (
    df_summary["total_defect_qty"]
    / df_summary["total_inspeciton_qty"]
)

df_summary["defect_ppm"] = (
    df_summary["defect_rate"] * 1000000
)

df_summary
```

```python
#cell7

bins = [
    0,
    100,
    300,
    600,
    np.inf,
]

labels = [
    "良好",
    "注意",
    "要確認",
    "重大確認",
]
```

```python
#cell8
df_summary["risk_category"] = pd.cut(
    df_summary["defect_ppm"],
    bins=bins,
    labels=labels,
    include_lowest=True,
)
df_summary
```

```python
df_summary["check_comment"] = "通常監視"
df_summary
```

```python
df_summary.loc[
    df_summary["risk_category"] == "注意",
    "check_comment",
] = "推移確認"
df_summary

df_summary.loc[
    df_summary["risk_category"] == "要確認",
    "check_comment",
] = "原因確認"

df_summary.loc[
    df_summary["risk_category"] == "重大確認",
    "check_comment",
] = "至急確認"

df_summary
```

```python
df_need_check = df_summary[
    df_summary["risk_category"].isin(
        [
            "要確認",
            "重大確認"
        ]
    )
].copy()
df_need_check
```

```python
df_risk_count = df_summary["risk_category"].value_counts(
    sort=False
).reset_index()
df_risk_count

df_risk_count.columns = [
    "risk_category",
    "count",
]

df_risk_count
```

```python
df_report = df_summary.sort_values(
    [
        "risk_category",
        "defect_ppm",
    ],
    ascending=[
        True,
        False,
    ],
)
df_report
```
