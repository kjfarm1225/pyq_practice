```python
#cell 3
df_work = df.copy()
df_work
```

```python
df_work["occ_date"] = pd.to_datetime(
    df_work["occ_date"]
)

df_work
```

```python
df_work["occ_date"].dtypes
```

```python
#cell5
df_summary = df_work.groupby(
    [
        "site",
        "defect_category",
    ],
    as_index=False,
)[
    [
        "defect_qty",
        "inspection_qty",
    ]
].sum()

df_summary
```

```python
#cell6
df_summary = df_summary.rename(
    columns={
        "defect_qty": "total_defect_qty",
        "inspection_qty": "total_inspection_qty",
    }
)
df_summary
```

```python
#cell7
df_summary["defect_rate"] = (
    df_summary["total_defect_qty"]
    / df_summary["total_inspection_qty"]
)

df_summary
```

```python
#cell8
df_summary["defect_ppm"] = (
    df_summary["defect_rate"] * 1000000
)
df_summary
```

```python
top3_index = df_summary["total_defect_qty"].nlargest(3).index
top3_index
```

```python
df_summary.loc[
    top3_index,
    "priority_rank",
] = [
    1,
    2,
    3,
]

df_summary
```

```python
#cell11
df_summary["priority_flag"] = "通常確認"
df_summary
```

```python
df_summary.loc[
    df_summary["priority_rank"].notna(),
    "priority_flag",
] = "重点確認"

df_summary
```

```python
df_summary["report_comment"] = "通常監視"

df_summary.loc[
    df_summary["priority_flag"] == "重点確認",
    "report_comment",
] = "不良数top3のための優先確認"
df_summary
```

```python
df_report = df_summary.sort_values(
    [
        "priority_rank",
        "total_defect_qty",
        "defect_ppm",
    ],
    ascending=[
        True,
        False,
        False,
    ],
)

df_report
```

```python
#cell14
df_priority = df_report[
    df_report["priority_flag"] == "重点確認"
].copy()

df_priority
```

```python
df_count_summary = pd.DataFrame(
    [
        ["集計後の組み合わせ数", len(df_summary)],
        ["重点確認件数", len(df_priority)],
        ["通常確認件数", len(df_summary) - len(df_priority)]
    ],
    columns=[
        "確認項目",
        "件数"
    ],
)

df_count_summary
```
