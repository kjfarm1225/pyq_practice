```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = "Meiryo"
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_csv(
    "dataset/daily_work_minutes_practice.csv",
)
df
```

```python
df["work_date_raw"] = df["work_date"]
df
```

```python
df["work_date_clean"] = (
    df["work_date"]
    .astype("string")
    .str.strip()
)
df
```

```python
df["work_date_clean"].dtype
```

```python
df["work_date_dt"] = pd.to_datetime(
    df["work_date_clean"],
    errors="coerce",
    format="mixed",
)
df
```

```python
df["work_date_dt"].dtypes
```

```python
df["task_type_clean"] = (
    df["task_type"]
    .astype("string")
    .str.strip()
)
df.head(3)
```

```python
df["work_minutes_clean"] = (
    df["work_minutes"]
    .astype("string")
    .str.strip()
)
df.head()
```

```python
df["work_minutes_num"] = pd.to_numeric(
    df["work_minutes_clean"],
    errors="coerce",
)
df
```

```python
df["work_minutes_clean"].dtypes
```

```python
df_data_ng = df[
    df["work_date_dt"].isna()
].copy()
df_data_ng
```

```python
df_minutes_ng = df[
    df["work_minutes_num"].isna()
].copy()

df_minutes_ng
```

```python
df_work = df[
    df["work_date_dt"].notna()
    & df["work_minutes_num"].notna()
    & df["task_type_clean"].notna()
].copy()
df_work
```

```python
df_work["work_date_day"] = df["work_date_dt"].dt.normalize()
df_work
```

```python
df_daily_task = (
    df_work
    .groupby(
        [
            "work_date_day",
            "task_type_clean",
        ]
    )
    .agg(
        work_minutes_sum=("work_minutes_num", "sum"),
        record_count=("work_minutes_num", "count"),
    )
)
df_daily_task
```

```python
date_index = pd.date_range(
    start=df_work["work_date_day"].min(),
    end=df_work["work_date_day"].max(),
    freq="D",
)
date_index
```

```python
task_types = df_work["task_type_clean"].drop_duplicates().tolist()
task_types
```

```python
full_index = pd.MultiIndex.from_product(
    [
        date_index,
        task_types,
    ],
    names=[
        "work_date",
        "task_type",
    ],
)
full_index
```

```python
filled_long = df_daily_task.reindex(
    full_index,
    fill_value=0,
)
filled_long
```

```python
filled_long = filled_long.reset_index()
filled_long
```

```python
matrix_df = filled_long.pivot(
    index="work_date",
    columns="task_type",
    values="work_minutes_sum",
)
matrix_df
```

```python
matrix_df = matrix_df.fillna(0)
matrix_df
```

```python
matrix_df.columns.name = None
matrix_df
```

```python
matrix_df["total_minutes"] = matrix_df[task_types].sum(axis=1)
```

```python
matrix_df
```

```python
graph_df = matrix_df[task_types].copy()
graph_df
```

```python
graph_df.index = graph_df.index.strftime("%Y-%m-%d")
graph_df
```

```python
graph_df.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 4),
)

plt.title("日別×作業分類別の作業時間")
plt.xlabel("日付")
plt.ylabel("作業時間（分）")
plt.xticks(rotation=45)
plt.legend(title="作業分類")
plt.tight_layout()
plt.show()
```
