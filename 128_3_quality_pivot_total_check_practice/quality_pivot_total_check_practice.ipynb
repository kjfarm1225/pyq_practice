```python
import pandas as pd
```

```python
df_total_sort = pd.DataFrame(
    [
        [12, 32, 9, 53],
        [18, 25, 7, 50],
        [8, 11, 4, 23],
    ],
    columns=[
        "melmb",
        "wse",
        "gse",
        "total",
    ],
    index=[
        "LCM-B",
        "LCM-A",
        "LCM-D",
    ],
)
df_total_sort
```

```python
df_total_sort["check_level"] = "通常確認"
df_total_sort
```

```python
df_total_sort.loc[df_total_sort["total"] >= 40, "check_level"] = "重点確認"
df_total_sort
```

```python
df_total_sort["comment"] = "通常監視"
df_total_sort
```
