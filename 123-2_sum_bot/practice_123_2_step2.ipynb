```python
def sum_bot():
    while s := input("> "):
        try:
            i = int(s)
            print("", i, "", sum(range(i + 1)), "です")
        except ValueError:
            print("数字を入力してください")
sum_bot()
```

```python
def double_bot():
    while s := input("> "):
        try:
            i = int(s)
            print(i * 2)
        except ValueError:
            print("数字を入力してください。")
double_bot()
```
