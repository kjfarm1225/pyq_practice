```python
import pandas as pd

df1 = pd.DataFrame(
    {
        "product": ["A", "B"],
        "prev": [100, 110],
    }
)

df2 = pd.DataFrame(
    {
        "product": ["A", "B"],
        "sales": [50, 120],
    }
)
_df1 = df1.rename(columns={"prev": "sales"})
print(df1)
print(_df1)

df_result = pd.concat([_df1, df2], ignore_index=True)
print(df_result)
```
