```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.read_csv("dataset/defect_comment_filter.csv")
df_defect
```

```python
#cell4
exclude_pattern = "TEST|確認用|サンプル|重複"
```

```python
#cell5
df_defect["exclude_flag"] = df_defect["comment"].str.contains(
    exclude_pattern,
    case=False,
    na=False,
)
df_defect
```

```python
df_actual = df_defect[
    ~df_defect["exclude_flag"]
].copy()
df_actual
```

```python
df_exclude = df_defect[
    df_defect["exclude_flag"]
].copy()
df_exclude
```

```python
df_site_count = df_actual.groupby(
    "site",
    as_index=False,
)["defect_id"].count()

df_site_count = df_site_count.rename(
    columns={
        "defect_id": "actual_count",
    }
)

df_site_count
```

```python
df_category_count = df_actual.groupby(
    "defect_category",
    as_index=False,
)["defect_id"].count()
df_category_count

df_category_count = df_category_count.rename(
    columns={
        "defect_id": "count"
    }
)
df_category_count
```
