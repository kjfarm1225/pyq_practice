```python
import pandas as pd
```

```python
index = pd.Index(
    [
        "LCM-A",
        "LCM-B",
        "LCM-C",
        "LCM-D",
    ],
    name="product",
)

columns = pd.Index(
    [
        "melmb",
        "wse",
        "gse",
    ],
    name="site",
)

data = [
    [18, 25, 7],
    [12, 32, 9],
    [5, 14, 31],
    [8, 11, 4],
]

df_wide = pd.DataFrame(
    data,
    index=index,
    columns=columns,
)

df_wide
```

```python
df_long = df_wide.melt(
    value_name="defect_qty",
    ignore_index=False
)

print(df_long)
df_long = df_long.reset_index()
df_long
```

```python
df_long_sort = df_long.sort_values(
    "defect_qty",
    ascending=False,
)
df_long_sort
```

```python
df_long_sort["check_level"] = "通常確認"
#df_long_sort

df_long_sort.loc[
    df_long_sort["defect_qty"] >= 20,
    "check_level",
] = "重点確認"

df_long_sort
```

```python
df_priority = df_long_sort[
    df_long_sort["check_level"] == "重点確認"
]

df_priority
```
