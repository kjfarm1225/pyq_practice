```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
y_temps = [28, 27, 27, 26, 26, 27, 27, 29, 30, 31, 33, 34,
           35, 35, 36, 35, 34, 33, 31, 30, 30, 29, 29, 29]
```

## オブジェクト指向インタフェース

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps_max = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
y_temps_min = [2.4, 6.2, 7.5, 14.3, 15.7, 20.3]

y_ticks = range(0, 31, 2)
x_ticks_labels = [f"{x+1}月" for x in range(6)]
```

```python
fig, ax = plt.subplots(figsize=(7, 3))

ax.plot(x_months, y_temps_max, label="最高気温",
         color="red",
         marker=".", markersize=10)

ax.plot(x_months, y_temps_min, label="最低気温",
         color="blue",
          marker="*", markersize=10,
           linestyle="--")

ax.grid(True, linestyle="--")
ax.set_yticks(y_ticks)
ax.set_xticks(x_months,
              labels=x_ticks_labels)

fig.suptitle("月別の最高・最低気温")
ax.set_xlabel("月")
ax.set_ylabel("気温（℃）")
ax.legend(loc="upper left")
plt.show()
```

## pyplotインターフェース

```python
plt.figure(figsize=(7, 3))

plt.plot(x_months, y_temps_max, label="最高気温",
          color="red",
           marker=".", markersize=10)
plt.plot(x_months, y_temps_min, label="最低気温",
          color="blue",
           marker="*", markersize=10,
            linestyle="--")

plt.grid(True, linestyle="--")
plt.yticks(y_ticks)
plt.xticks(x_months,
            labels=x_ticks_labels)

plt.suptitle("月別の最高・最低気温")
plt.xlabel("月")
plt.ylabel("気温（℃）")
plt.legend(loc="lower right")

plt.show()
```
