## 1問目 グラフ画像の保存

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
x_hours = range(0, 24)
```

```python
y_temps = [28, 27, 27, 26, 26, 27, 27, 29, 30, 31, 33, 34,
           35, 35, 36, 35, 34, 33, 31, 30, 30, 29, 29, 29]
```

```python
fig, ax = plt.subplots()
ax.plot(x_hours, y_temps, marker="s")

fig.suptitle("時間帯ごとの気温")
ax.set_xlabel("時間帯（時）")
ax.set_ylabel("気温")
ax.set_xticks(range(0, 24, 3))
ax.grid(True)

fig.savefig("sample.png")
```
