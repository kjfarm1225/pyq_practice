## 1問目 グラフ画像の保存

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
x_hours = range(0, 24)
```

```python
y_temps = [28, 27, 27, 26, 26, 27, 27, 29, 30, 31, 33, 34,
           35, 35, 36, 35, 34, 33, 31, 30, 30, 29, 29, 29]
```

```python
fig, ax = plt.subplots()
ax.plot(x_hours, y_temps, marker="s")

fig.suptitle("時間帯ごとの気温")
ax.set_xlabel("時間帯（時）")
ax.set_ylabel("気温")
ax.set_xticks(range(0, 24, 3))
ax.grid(True)

fig.savefig("sample.png")
```

## 演習問題

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
months = [1, 2, 3, 4, 5, 6]
defect_counts = [4, 6, 3, 8, 5, 7]
```

```python
fig, ax = plt.subplots()
ax.plot(months, defect_counts, marker="o")

fig.suptitle("月別の不具合件数")

ax.set_xlabel("月")
ax.set_ylabel("不具合件数（件）")
ax.grid(True)

plt.show()

fig.savefig("defect_sample.png")
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rc("font", family="Meiryo")
```

```python
months =[1, 2, 3, 4, 5, 6]
defect_counts = [4, 6, 3, 8, 5, 7]
```

```python
fig, ax = plt.subplots()

ax.plot(months, defect_counts, marker="s")

fig.suptitle("上期の不具合件数")
ax.set_xlabel("月")
ax.set_ylabel("不具合件数（件）")
ax.grid(True)

plt.show()

fig.savefig("defect_report.png")
```
