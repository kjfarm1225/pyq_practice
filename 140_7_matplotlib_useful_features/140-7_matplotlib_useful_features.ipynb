## 140-7「Matplotlibの便利な機能」2問目

## オブジェクト指向インタフェース

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps_max = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
y_temps_min = [2.4, 6.2, 7.5, 14.3, 15.7, 20.3]

y_ticks = range(0, 31, 2)
x_ticks_labels = [f"{x+1}月" for x in range(6)]
```

```python
fig, ax = plt.subplots(figsize=(7, 3))

ax.plot(x_months, y_temps_max, label="最高気温",
         color="red",
         marker=".", markersize=10)

ax.plot(x_months, y_temps_min, label="最低気温",
         color="blue",
          marker="*", markersize=10,
           linestyle="--")

ax.grid(True, linestyle="--")
ax.set_yticks(y_ticks)
ax.set_xticks(x_months,
              labels=x_ticks_labels)

fig.suptitle("月別の最高・最低気温")
ax.set_xlabel("月")
ax.set_ylabel("気温（℃）")
ax.legend(loc="upper left")
plt.show()
```

## pyplotインターフェース

```python
plt.figure(figsize=(7, 3))

plt.plot(x_months, y_temps_max, label="最高気温",
          color="red",
           marker=".", markersize=10)
plt.plot(x_months, y_temps_min, label="最低気温",
          color="blue",
           marker="*", markersize=10,
            linestyle="--")

plt.grid(True, linestyle="--")
plt.yticks(y_ticks)
plt.xticks(x_months,
            labels=x_ticks_labels)

plt.suptitle("月別の最高・最低気温")
plt.xlabel("月")
plt.ylabel("気温（℃）")
plt.legend(loc="lower right")

plt.show()
```

```python
fig, ax1 = plt.subplots()

ax2 = ax1.twinx()

ax1.bar(months, production_counts, color="skyblue")
ax1.set_xlabel("月", size=15)
ax1.set_ylabel("生産数（個）", size=15)

ax2. plot(months, ppm_values, color="red", marker="o")
ax2.set_ylabel("不良率(ppm)", size=15)

fig.suptitle("月別の生産数と不良率ppm", size=20)
plt.show()
```

## 実務練習：月別の生産数・不具合件数・不良率ppmを2軸グラフで可視化する

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"

months = [1, 2, 3, 4, 5, 6]
month_labels = ["1月", "2月", "3月", "4月", "5月", "6月"]

production_counts = [1200, 1350, 1280, 1420, 1500, 1600]
defect_counts = [4, 6, 3, 8, 5, 7]

ppm_values = [defect / production * 1000000
              for defect, production in zip(defect_counts, production_counts)]
```

```python
fig, ax1 = plt.subplots(figsize=(8,4))
ax2 = ax1.twinx()

ax1.bar(months, production_counts, color="skyblue", label="生産数")
ax1.set_xlabel("月", size=13)
ax1.set_ylabel("生産数（個）", size=13)
ax1.set_xticks(months, labels=month_labels)
ax1.grid(True, axis="y", linestyle="--")
ax1.set_axisbelow(True)

ax2.plot(months, ppm_values, color="red", marker="o", label="不良率ppm")
ax2.set_ylabel("不良率（ppm）", size=13)

for month, defect, ppm in zip(months, defect_counts, ppm_values):
    ax2.text(month, ppm + 100, f"{defect}件", ha="center", color="red")

fig.suptitle("月別の生産数と不良率ppm", size=18)

ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(8,4))
ax2 = ax1.twinx()

ax1.bar(months, production_counts, color="skyblue", label="生産数")
ax1.set_xlabel("月", size=13)
ax1.set_ylabel("生産数（個）", size=13)
ax1.set_xticks(months, labels=month_labels)
ax1.grid(True, axis="y", linestyle="--")
ax1.set_axisbelow(True)

ax2.plot(months, ppm_values, color="red", marker="o", label="不良率ppm")
ax2.set_ylabel("不良率（ppm）", size=13)

for month, defect, ppm in zip(months, defect_counts, ppm_values):
    ax2.text(month, ppm + 100, f"{defect}件", ha="center", color="red")

fig.suptitle("月別の生産数と不良率ppm", size=18)

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(handles1 + handles2,
           labels1 + labels2,
           loc="upper left")

plt.show()
```
