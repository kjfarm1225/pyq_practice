# 140-6 複数のサブプロット
## 1問目 複数のサブプロットの作成

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"

x_months = [1, 2, 3, 4, 5, 6]
y_temps = [5.2, 7.8, 12.4, 18.9, 22.1, 26.3]
y_rainfall = [65.5, 72.3, 120.2, 131.4, 145.4, 320.8]
```

```python
fig, axes = plt.subplots(2)

axes[0].plot(x_months, y_temps)
axes[1].plot(x_months, y_rainfall)

fig.suptitle("ある地域の月別の平均気温・合計降水量")

plt.show()
```

```python
fig, axes = plt.subplots(1, 2)
axes[0].plot(x_months, y_temps)
axes[1].bar(x_months, y_rainfall)

fig.suptitle("ある地域の月別の平均気温・合計降水量")
plt.show()
```

## 実務的な簡単写経練習

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"

months = [1, 2, 3, 4, 5, 6]
defect_counts = [4, 6, 3, 8, 5, 7]
production_counts = [1200, 1350, 1280, 1420, 1500, 1600]
```

```python
fig, axes = plt.subplots(2)
axes[0].plot(months, defect_counts, marker="o")
axes[1].bar(months, production_counts)

fig.suptitle("月別の不具合件数と生産数")
plt.show()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"

months = [1, 2, 3, 4, 5, 6]
defect_counts = [4, 6, 3, 8, 5, 7]
production_counts = [1200, 1350, 1280, 1420, 1500, 1600]
```

```python
fig, axes = plt.subplots(2)

axes[0].plot(months, defect_counts, marker="o")
axes[0].set_title("月別不具合件数")
axes[0].set_ylabel("不具合件数（件）")
```
