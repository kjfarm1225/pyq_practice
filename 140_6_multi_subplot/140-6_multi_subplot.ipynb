## 実務練習 複数のサブプロットを見やすく整える

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
months = [1, 2, 3, 4, 5, 6]
month_labels =["1月", "2月", "3月", "4月", "5月", "6月"]

defect_counts = [4, 6, 3, 8, 5, 7]
production_counts = [1200, 1350, 1280, 1420, 1500, 1600]
```

```python
fig, axes = plt.subplots(2, figsize=(7, 5), layout="constrained")

axes[0].plot(months, defect_counts, marker="o")
axes[0].set_title("月別不具合件数")
axes[0].set_xlabel("月")
axes[0].set_ylabel("不具合件数(件)")
axes[0].set_xticks(months, month_labels)
axes[0].grid()

axes[1].bar(months, production_counts, zorder=3)
axes[1].set_title("月別生産数")
axes[1].set_xlabel("月")
axes[1].set_ylabel("生産数（個）")
axes[1].set_xticks(months, month_labels)
axes[1].set_axisbelow(True)
axes[1].grid()

plt.show()
```
