# 140-6 複数のサブプロット
## 1問目 複数のサブプロットの作成

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"

x_months = [1, 2, 3, 4, 5, 6]
y_temps = [5.2, 7.8, 12.4, 18.9, 22.1, 26.3]
y_rainfall = [65.5, 72.3, 120.2, 131.4, 145.4, 320.8]
```

```python
fig, axes = plt.subplots(2)

axes[0].plot(x_months, y_temps)
axes[1].plot(x_months, y_rainfall)

fig.suptitle("ある地域の月別の平均気温・合計降水量")

plt.show()
```
