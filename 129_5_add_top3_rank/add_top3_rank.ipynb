###### 129_5_add_top3_rank

```python
#cell1
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        ["Alice", "国語", 92],
        ["Alice", "数学", 72],
        ["Bob", "国語", 65],
        ["Bob", "数学", 92],
    ],
    columns=[
        "Name",
        "Subject",
        "Point",
    ],
)
df
```

```python
#cell3
top3_index = df["Point"].nlargest(3).index
top3_index
```

```python
df.loc[
    top3_index,
    "Rank",
] = [
    1,
    2,
    3,
]
df
```
