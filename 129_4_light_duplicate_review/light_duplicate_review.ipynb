```python
#cell 1
import pandas as pd
```

```python
#cell 2
df = pd.DataFrame(
    [
         ["S001", "J69P", "MELMB", "OK", "2026-06-02 09:00"],
        ["S002", "J69P", "WSE", "NG", "2026-06-02 09:10"],
        ["S003", "310D", "GSE", "OK", "2026-06-02 09:20"],
        ["S002", "J69P", "WSE", "OK", "2026-06-02 10:00"],
        ["S004", "310D", "MELMB", "OK", "2026-06-02 10:10"],
        ["S005", "J69P", "GSE", "NG", "2026-06-02 10:20"],
        ["S005", "J69P", "GSE", "OK", "2026-06-02 11:00"],
    ],
    columns=[
        "serial_no",
        "project",
        "site",
        "inspection_result",
        "update_time",
    ],
)
df
```

```python
#cell3
df_work = df.copy()
df_work
```

```python
#cell4
df_work["update_time"] = pd.to_datetime(
    df_work["update_time"]
)

df_work
df_work["update_time"].dtype
```

```python
df_duplicate_check = df_work[
    df_work.duplicated(
        "serial_no",
        keep=False,
    )
].copy
df_duplicate_check
```

```python
#cell 6
before_count = len(df_work)
before_count
```

```python
df_sort = df_work.sort_values(
    "update_time",
)
df_sort
```

```python
df_result = df_sort.drop_duplicates(
    "serial_no",
    keep="last",
)
df_result
```

```python
after_count = len(df_result)
after_count
```

```python
#cell 10
deleted_count = before_count - after_count
deleted_count
```

```python
#cell 11
df_result_cross = df_result.pivot_table(
    values="serial_no",
    index="site",
    columns="inspection_result",
    aggfunc="count",
    fill_value=0,
)
df_result_cross
```
