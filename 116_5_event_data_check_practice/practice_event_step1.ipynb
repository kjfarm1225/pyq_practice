```python
code_map = {"S": False, "L": True, "SL": True}
df["リモート可"] = df["開催形式"].map(code_map)
df.head()
```

```python
df["参加者数"] = df["登録者数"].mask(
    df["登録者数"] > df ["定員"],
    df["定員"]
)
df.head()
```

```python
df["満席率"] = df["登録者数"] / df["定員"]
df.head()
```

```python
def categorize(x):
    if "次週" in  x or "もくもく" in x:
        return "次週"
    elif "輪読" in x:
        return "輪読"
    else:
        return "一般"
df["カテゴリー"] = df["イベント"].apply(categorize)
df.head()
```
