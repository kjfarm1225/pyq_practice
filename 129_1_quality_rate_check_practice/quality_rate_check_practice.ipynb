```python
#cell1

import pandas as pd
```

```python

#cell2

df = pd.DataFrame(
    [
        ["Q001", "J69P", "MELMB", "Line1", 1200, 3],
        ["Q002", "J69P", "WSE", "Line2", 980, -2],
        ["Q003", "310D", "GSE", "Line1", -500, 1],
        ["Q004", "310D", "ISE", "Line2", 760, 0],
        ["Q005", "J69P", "MELMB", "Line1", 1100, 5],
        ["Q006", "310D", "WSE", "Line2", 850, -1],
        ["Q007", "J69P", "GSE", "Line1", 920, 2],
        ["Q008", "310D", "MELMB", "Line2", 0, 1],
        ["Q009", "J69P", "ISE", "Line1", 1500, 8],
        ["Q010", "310D", "GSE", "Line2", 1300, 2],
    ],
    columns=[
        "check_id",
        "project",
        "site",
        "line",
        "inspection_qty",
        "defect_qty",
    ],
)
df
```

```python
df["defect_qty_fixed"] = df["defect_qty"].where(
    df["defect_qty"] >= 0,
    0,
)
df
```

```python
df["inspection_qty_fixed"] = df["inspection_qty"].where(
    df["inspection_qty"] >= 0,
    0,
)
df
```

```python
df["defect_rate_safe"] = (
    df["defect_qty_fixed"] / df["inspection_qty_fixed"]
).where(
    df["inspection_qty_fixed"] > 0,
    0,
)
df
```

```python
df["check_result"] = "通常"

df["check_result"] = df["check_result"].where(
    df["defect_rate_safe"] <= 0.003,
    "要確認",
)
df
```
