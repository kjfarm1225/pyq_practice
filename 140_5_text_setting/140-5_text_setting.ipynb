## 実務練習1 月別不具合件数のピークにテキストを付ける

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
months = [1, 2, 3, 4, 5, 6]
defects = [3, 5, 4, 8, 6, 5]
```

```python
max_defects = max(defects)
max_month = months[defects.index(max_defects)]
```

```python
fig, ax = plt.subplots()

ax.plot(months, defects, marker="o")
ax.text(max_month, max_defects, "最多")

fig.suptitle("月別の不具合件数")
ax.set_xlabel("月")
ax.set_ylabel("不具合件数（件）")

ax.set_xticks(months)
ax.grid(True)

plt.show()
```

## 実務練習2 月別不具合件数にデータラベルを表示する

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
months = [1, 2, 3, 4, 5, 6]
defect_counts = [3, 5, 4, 8, 6, 5]
```

```python
fig, ax = plt.subplots()

ax.plot(months, defect_counts, marker="o")

for x, y  in zip(months, defect_counts):
    ax.text(x, y + 0.3, str(y),
            ha="center",
            backgroundcolor=(1, 1, 1, 0.4))
    
fig.suptitle("月別の不具合件数")
ax.set_xlabel("月")
ax.set_ylabel("不具合件数（件）")

ax.set_xticks(months)
ax.grid(True)

plt.show()
```

## 3問目 テキストとフォント ── 補足説明と文字サイズの総合演習

```python
import matplotlib.pyplot as plte
%matplotlib inline

plt.rcParams["font.family"] = "Meiryo"
```

```python
x_ticks = [1, 2, 3, 4]
x_ticklabels = ["6/3週", "6/10週", "6/17週", "6/24週"]

y_sales = [234, 262, 352, 281]
```

```python
fig, ax = plt.subplots()

ax.plot(x_ticks, y_sales, marker="s")

ax.text(2.5, 360,
        "キャンペーン期間",
        color="red",
        size=20)
fig.suptitle("週ごとの売上（6月分）",
             size=20)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

# 日本語フォントの設定
plt.rcParams["font.family"] = "Meiryo"

# X軸の値: 週
x_ticks = [1, 2, 3, 4]  # X軸の目盛の位置
x_ticklabels = ["6/3週", "6/10週", "6/17週", "6/23週"]  # X軸の目盛ラベル

# Y軸の値: 各週の売上
y_sales = [234, 262, 352, 281]

fig, ax = plt.subplots()  # 描画オブジェクト（fig）とサブプロット（ax）を作成

# サブプロットに、折れ線グラフを追加
ax.plot(x_ticks, y_sales, marker="s")

# X座標が2.5、Y座標が360の位置に「キャンペーン期間」と描画する
ax.text(2.5, 360,
        "キャンペーン期間",
        color="red",
        size=12)

# タイトルと軸ラベルの設定
fig.suptitle("週ごとの売上（6月分）",
             size=20)

ax.set_xlabel("週",
              size=15)

ax.set_ylabel("売上（万円）",
              size=15)

ax.set_ylim(200, 380)  # Y軸の範囲
ax.set_xticks(x_ticks, x_ticklabels)  # X軸の目盛位置とラベル
ax.grid(True)  # グリッド線を表示

plt.show()  # 図全体を表示
```
