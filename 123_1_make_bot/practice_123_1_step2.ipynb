```python
def repeat_bot():
    s  = input("> ")

    while s != "":
        print("あなたが入力したのは", s, "です")
        s = input("> ")

repeat_bot()
```

```python
def echo_bot():
    while s := input("> "):
        print("あなたが入力したのは", s, "です")

echo_bot()
```
