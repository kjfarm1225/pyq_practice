```python
import pandas as pd
import numpy as np
```

```python
#cell2
df = pd.DataFrame(
    [
        ["D001", "J69P", "MELMB", "コイル", 3, 12000, "調査中"],
        ["D002", "J69P", "WSE", "Buck-IC", 5, 18000, "対策検討中"],
        ["D003", "310D", "GSE", "コイル", 12, 25000, "暫定対策済"],
        ["D004", "310D", "MELMB", "はんだ", 2, 9000, "調査中"],
        ["D005", "J69P", "GSE", "コイル", 7, 15000, "調査中"],
        ["D006", "310D", "WSE", "コネクタ", 1, 6000, "完了"],
        ["D007", "J69P", "MELMB", "Buck-IC", 8, 20000, "調査中"],
        ["D008", "310D", "GSE", "コイル", 4, 10000, "対策検討中"],
        ["D009", "J69P", "GSE", "Buck-IC", 6, 14000, "調査中"],
        ["D010", "310D", "MELMB", "コイル", 5, 16000, "暫定対策済"],
        ["D011", "J69P", "WSE", "コイル", 9, 22000, "調査中"],
        ["D012", "310D", "GSE", "はんだ", 3, 8000, "完了"],
    ],
    columns=[
            "defect_id",
            "project",
            "site",
            "defect_category",
            "defect_qty",
            "inspection_qty",
            "status",
    ],
)
df
```

```python
df_work = df.copy()
df_work
```

```python
bins = [
    0,
    10000,
    20000,
    np.inf,
]

labels = [
    "少ない",
    "普通",
    "多い",
]
```

```python
#cell5
df_work["inspection_category"] = pd.cut(
    df_work["inspection_qty"],
    bins=bins,
    labels=labels,
    include_lowest=True,
)
df_work
```

```python
#cell6
df_inspection_count = df_work["inspection_category"].value_counts(
    sort=False
).reset_index()
df_inspection_count

df_inspection_count.colums = [
    "inspection_category",
    "count",
]

df_inspection_count
```

```python
df_many_inspection = df_work[
    df_work["inspection_category"] == "多い"
].copy()
df_many_inspection
```

```python
df_work["inspection_comment"] = "通常確認"
df_work

df_work.loc[
    df_work["inspection_category"] == "少ない",
    "inspection_comment",
] = "サンプル数少ないため参考確認"
df_work

df_work.loc[
    df_work["inspection_category"] == "多い",
    "inspection_comment",
] = "検査数多めのため傾向確認しやすい"

df_work
```

```python
df_report = df_work.sort_values(
    "inspection_qty",
    ascending=False,
)
df_report
```
