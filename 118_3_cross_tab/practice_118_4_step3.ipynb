```python
import pandas as pd

df_log = pd.read_csv("dataset/product_log.csv")
df_log.head()
```

```python
df_log["Month"] = df_log["Date"].str[5:7].astype(int)
df_log.head()
```

```python
df_log_pivot = df_log.pivot_table("Num", "Product", "Month", "sum")
df_log_pivot
```

```python
df_log_pivot = df_log.pivot_table(
    values="Num",
    index="Product",
    columns="Month",
    aggfunc="sum"    
)
df_log_pivot
```

```python
df_log_pivot = df_log.pivot_table(
    values="Num",
    index="Product",
    columns="Month",
    aggfunc="mean"
)
df_log_pivot
```
