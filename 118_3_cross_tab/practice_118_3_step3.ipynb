```python
import pandas as pd
%matplotlib inline

df_log = pd.read_csv("dataset/product_log.csv")
df_log["Month"] = df_log["Date"].str[5:7].astype(int)

df_log_pivot = df_log.pivot_table(
    values="Num",
    index="Product",
    columns="Month",
    aggfunc="sum"
)
df_log_pivot.plot()
df_log_pivot.T
df_log_pivot.T.plot()
```

```python
df_log_pivot.plot.bar()
```

```python
df_log_pivot.T.plot.bar()
```

```python
ax = df_log_pivot.T.plot()
ax.set_title("Monthly Sales by product")
ax.set_xlabel("Month")
ax.set_ylabel("Sales Num")
```
