```python
import pandas as pd
```

```python
df_defect = pd.DataFrame(
    [
        ["2026-06", "MELMB", "LCM-A", "点灯不良", 18],
        ["2026-06", "WSE", "LCM-A", "点灯不良", 25],
        ["2026-06", "GSE", "LCM-A", "点灯不良", 7],
        ["2026-06", "MELMB", "LCM-B", "ショート", 12],
        ["2026-06", "WSE", "LCM-B", "ショート", 32],
        ["2026-06", "GSE", "LCM-B", "ショート", 9],
        ["2026-06", "MELMB", "LCM-C", "外観キズ", 5],
        ["2026-06", "WSE", "LCM-C", "外観キズ", 14],
        ["2026-06", "GSE", "LCM-C", "外観キズ", 21],
        ["2026-06", "MELMB", "LCM-D", "コネクタ浮き", 8],
        ["2026-06", "WSE", "LCM-D", "コネクタ浮き", 11],
        ["2026-06", "GSE", "LCM-D", "コネクタ浮き", 4],
    ],
    columns=[
        "month",
        "site",
        "product",
        "defect_type",
        "defect_qty",
    ],
)

df_defect.head()
```

```python
df_product_master = pd.DataFrame(
    [
        ["LCM-A", "ヘッドランプ", "品質保証1課"],
        ["LCM-B", "ヘッドランプ", "品質保証1課"],
        ["LCM-C", "リアランプ", "品質保証2課"],
        ["LCM-D", "ヘッドランプ", "品質保証1課"],
    ],
    columns=[
        "product",
        "category",
        "owner_department",
    ],
)
df_product_master
```

```python
df_merge = df_defect.merge(df_product_master, on="product")
df_merge.tail()
```

```python
df_headlamp = df_merge[df_merge["category"] == "ヘッドランプ"]
df_headlamp
```

```python
df_wide = df_headlamp.pivot(
    index="product",
    columns="site",
    values="defect_qty",
)
df_wide
```

```python
df_wide = df_wide[["MELMB", "WSE", "GSE"]]
df_wide
```

```python
df_wide["total"] = df_wide["MELMB"] + df_wide["WSE"] + df_wide["GSE"]
df_wide
```

```python
df_total_sort = df_wide.sort_values("total", ascending=False)
df_total_sort
```
