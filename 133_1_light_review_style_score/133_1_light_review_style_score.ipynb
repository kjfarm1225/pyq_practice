```python
import pandas as pd
```

```python
df = pd.read_csv(
    "dataset/score.csv",
)
df
```

```python
score_cols = ["国語", "数学", "理科", "社会"]
```

```python
def style_score(x):
    if x < 70:
        style = "font-weight: bold;"
    elif x >= 90:
        style = "background-color: yellow;"
    else:
        style = None
    return style
```

```python
styled_df = df.style.map(
    style_score,
    subset=score_cols
)
styled_df
```
