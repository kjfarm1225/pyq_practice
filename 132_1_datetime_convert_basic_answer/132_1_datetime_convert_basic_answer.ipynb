```python
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("dataset/study_log_dirty.csv")
df
```

```python
df.dtypes
```

```python
df["start_at"] = pd.to_datetime(
    df["start_at"],
    format="mixed",
)
df
```

```python
df["end_at"] = pd.to_datetime(
    df["end_at"],
    format="mixed",
)
df
```

```python
df.dtypes
```

```python
df["start_at"] = pd.to_datetime(
    df["start_at"],
    format="mixed",
    errors="coerce",
)
```

```python
df["end_at"] = pd.to_datetime(
    df["end_at"],
    format="mixed",
    errors="corece",
)
```

```python
df["study_minutes"] = (
    df["end_at"] - df["start_at"]
).dt.total_seconds() / 60
```

```python
df
```

```python
df["study_minutes"] = df["study_minutes"].round(0).astype(int)
```

```python
df
```

```python
df.drop(columns=["stury_minutes"], inplace=True)
```

```python
df
```

```python
df.columns
```

```python
df["start_date"] = df["start_at"].dt.date
```

```python
df
```

```python
df["start_month"] = df["start_at"].dt.to_period("M").astype(str)
```

```python
df
```

```python
df_summary = df.groupby(
    ["start_month", "subject"],
    as_index=False,
)["study_minutes"].sum()
```
