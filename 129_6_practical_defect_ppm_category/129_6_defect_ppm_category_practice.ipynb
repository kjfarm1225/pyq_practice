```python
#cell1
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(
    [
        ["D001", "J69P", "MELMB", "コイル", 3, 12000, "調査中"],
        ["D002", "J69P", "WSE", "Buck-IC", 5, 18000, "対策検討中"],
        ["D003", "310D", "GSE", "コイル", 12, 25000, "暫定対策済"],
        ["D004", "310D", "MELMB", "はんだ", 2, 9000, "調査中"],
        ["D005", "J69P", "GSE", "コイル", 7, 15000, "調査中"],
        ["D006", "310D", "WSE", "コネクタ", 1, 6000, "完了"],
        ["D007", "J69P", "MELMB", "Buck-IC", 8, 20000, "調査中"],
        ["D008", "310D", "GSE", "コイル", 4, 10000, "対策検討中"],
        ["D009", "J69P", "GSE", "Buck-IC", 6, 14000, "調査中"],
        ["D010", "310D", "MELMB", "コイル", 5, 16000, "暫定対策済"],
        ["D011", "J69P", "WSE", "コイル", 9, 22000, "調査中"],
        ["D012", "310D", "GSE", "はんだ", 3, 8000, "完了"],
    ],
    columns=[
        "defect_id",
        "project",
        "site",
        "defect_category",
        "defect_qty",
        "inspection_qty",
        "status",
    ],
)
df
```

```python
#cell3
df_work = df.copy()
df_work
```

```python
df_work["defect_rate"] =(
    df_work["defect_qty"]
    / df_work["inspection_qty"]
)
df_work
```

```python
df_work["defect_ppm"] = (
    df_work["defect_rate"] * 1000000
)
df_work
```

```python
bins = [
    0,
    100,
    300,
    600,
    np.inf,
]
labels =[
    "良好",
    "注意",
    "要確認",
    "重大確認",
]
```

```python
#cell7
df_work["risk_category"] = pd.cut(
    df_work["defect_ppm"],
    bins=bins,
    labels=labels,
    include_lowest=True,
)
df_work
```

```python
df_serious = df_work[
    df_work["risk_category"] == "重大確認"
].copy()

df_serious
```

```python
#cell9
df_risk_count = df_work["risk_category"].value_counts(
    sort=False
).reset_index()

df_risk_count.columns= [
    "risk_category",
    "count",
]

df_risk_count
```

```python
df_work["report_comment"] = "通常監視"
df_work

df_work.loc[
    df_work["risk_category"] == "注意",
    "report_comment",
] = "推移確認"
df_work

df_work.loc[
    df_work["risk_category"] == "要確認",
    "report_comment",
] = "原因確認"
df_work

df_work.loc[
    df_work["risk_category"] == "重大確認",
    "report_comment",
] = "至急確認"
df_work
```

```python
#cell11

df_report = df_work.sort_values(
    [
        "risk_category",
        "defect_ppm",
    ],
    ascending=[
        True,
        False,
    ],
)
df_report
```
