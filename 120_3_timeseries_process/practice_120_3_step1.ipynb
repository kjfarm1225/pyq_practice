```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

df_sale = pd.read_csv(
    "dataset/product_sale.csv",
    parse_dates=["Date"],
    index_col="Date"
)

df_sale["Sale"].plot(marker="o")
plt.legend();

df_sale[["Sale"]].plot(marker="o")
```

```python
#練習1
#Sale ではなく、Num 列のグラフを書いてみてください。
df_sale["Num"].plot(marker="o")
plt.legend();
#練習2
#Price 列のグラフを書いてみてください。
df_sale["Price"].plot(marker="o")
plt.legend();

#練習3
#marker="o" を marker="*" に変えて、見た目の違いを確認してください。
```

```python
#練習3
df_sale["Sale"].plot(marker="*")
plt.legend();
```
