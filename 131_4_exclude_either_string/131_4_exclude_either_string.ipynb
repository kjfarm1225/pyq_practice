```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/review.csv", index_col="user_id")
df
```

```python
contains_110_or_220 = df["comment"].str.contains("110|220")
contains_110_or_220
```

```python
selected_df = df[
    ~contains_110_or_220
]
selected_df
```
