```python
import pandas as pd
df = pd.read_csv("event_after_sample.csv")
df.head()
```

```python
grouped = df.groupby("勉強会グループ")
summary_df = grouped["リモート可"].agg(["size", "sum"])
summary_df = summary_df.rename(
    columns={
        "size": "イベント数",
        "sum": "リモート可のイベント数"
    }
)
summary_df["参加者の延べ人数"] = grouped["参加者数"].sum()

bins = [-float("inf"), 50, 100, 500, 1000, float("inf")]
summary_df["参加者の延べ人数の区分"] = pd.cut(
    summary_df["参加者の延べ人数"],
    bins,
    right=False
)
summary_df
```
