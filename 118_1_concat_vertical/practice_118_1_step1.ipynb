```python
import pandas as pd

df1 = pd.DataFrame(
    [['2017/1/3', 'A', 3], ['2017/2/4', 'B', 4]],
    columns=['Date', 'Name', 'Size']
)
df1
```

```python
df2 = pd.DataFrame(
    [['2017/3/5', 'A', 5]],
    columns=['Date', 'Name', 'Size']
)
df2
```

```python
df3 = pd.concat([df1,df2])
df3
```

```python
df3 = pd.concat([df1, df2], ignore_index=True)
df3
```

```python
df4 = pd.concat([df1, df2]).reset_index(drop=True)
df4
```

```python
df3 == df4
```

```python
(df3 == df4).all()
```

```python
(df3 == df4).all().all()
```

```python
df_a = pd.DataFrame(
    [['2024/1/1', 'X', 10], ['2024/1/2', 'Y', 20]],
    columns=['Date', 'Name', 'Size']
)

df_b = pd.DataFrame(
    [['2024/1/3', 'Z', 30]],
    columns=['Date', 'Name', 'Size']
)

df_c = pd.concat([df_a, df_b]).reset_index(drop=True)
df_c
```

```python
df_d = pd.concat([df_a, df_b]).reset_index(drop=True)
df_d
```

```python
(df_c == df_d).all().all()
```
