```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/study_log_naive.csv",
                 parse_dates=["start_at", "end_at"],)
```

```python
df
```

```python
df["start_at"]
```

```python
start_at_utc = df["start_at"].dt.tz_localize("UTC")
```

```python
start_at_utc
```

```python
df["start_at"] = start_at_utc.dt.tz_convert("Asia/tokyo")
```
