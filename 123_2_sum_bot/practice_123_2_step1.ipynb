```python
def sum_bot():
    s  = input("> ")

    while s != "":
        i = int(s)
        print(sum(range(i + 1)))
        s = input("> ")
    print("botを終了します。")
sum_bot()
```

```python
def sum_bot():
    while s := input("> "):
        i = int(s)
        print("1から", i, "までの合計は:", sum(range(i + 1)))

sum_bot()
```

```python
def double_bot():
    while s := input("> "):
        i = int(s)
        print(i * 2)
double_bot()
```
