```python
import pandas as pd
```

```python
df = pd.read_csv(
    "dataset/lcm_monthly_quality_format_practice.csv",
)
```

```python
display(df)
```

```python
df.dtypes
```

```python
df["inspection_qty"] = pd.to_numeric(
    df["inspection_qty_raw"],
    errors="coerce",
)
```

```python
df["inspection_qty"].dtypes
```

```python
df["inspection_qty_raw"].dtypes
```

```python
#cell 5 残りの列も数値化する。
df["defect_qty"] = pd.to_numeric(
    df["defect_qty_raw"],
    errors="coerce",
)

df["ppm"] = pd.to_numeric(
    df["ppm_raw"],
    errors="coerce",
)

df["change_rate"] = pd.to_numeric(
    df["change_rate_raw"],
    errors="coerce",
)
```

```python
df.dtypes
```

```python
#cell 6
mask_inspection_ng = (
    df["inspection_qty_raw"].notna()
    & df["inspection_qty"].isna()
)
```

```python
mask_defect_ng = (
    df["defect_qty_raw"].notna()
    & df["defect_qty"].isna()
)

mask_ppm_ng = (
    df["ppm_raw"].notna()
    & df["ppm"].isna()
)

mask_change_rate_ng = (
    df["change_rate_raw"].notna()
    & df["change_rate"].isna()
)
```

```python
#cell 8
mask_any_ng = (
    mask_inspection_ng
    | mask_defect_ng
    | mask_ppm_ng
    | mask_change_rate_ng
)
```

```python
#cell 9
df_ng = df[
    mask_any_ng
].copy()
display(df_ng)
```

```python
df_work  = df[
    ~mask_any_ng
].copy()
```

```python
df_work.dtypes
```

```python
df_display = df_work[
    [
        "site",
        "month",
        "inspection_qty",
        "defect_qty",
        "ppm",
        "change_rate",
    ]
].copy()
```

```python
display(df_display)
```

```python
display(df_display)
```

```python
display(
    df_display.style.format(
        {
            "inspection_qty": "{:,.0f}",
            "defect_aty": "{:.0f}",
            "ppm": "{:.1f}",
            "change_rate": "{:.1%}",
        },
        na_rep="-",
    )
)
```
