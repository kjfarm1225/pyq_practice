```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = "Meiryo"
plt.rcParams["axes.unicode_minus"] = False
```

```python
#cell 2
df = pd.read_csv(
    "dataset/lcm_weekly_defect_work_practice.csv",
)
df
```

```python
df["occurrence_date_raw"] = df["occurrence_date"]
df
```

```python
df["occurrence_date"].dtypes
```

```python
df["occurrence_date_clean"] = (
    df["occurrence_date"]
    .astype("string")
    .str.strip()
)
df
```

```python
#cell 6
df["occurrence_date_dt"] = pd.to_datetime(
    df["occurrence_date_clean"],
    errors="coerce",
    format="mixed",
)
df
```

```python
df["occurrence_date_dt"].dtypes
```

```python
df["site_clean"]  = (
    df["site"]
    .astype("string")
    .str.strip()
)
df
```

```python
#cell 8
df["defect_category_clean"] = (
    df["defect_category"]
    .astype("string")
    .str.strip()
)
df
```

```python
df["defect_count_clean"] = (
    df["defect_count"]
    .astype("string")
    .str.strip()
)
df.head()
```

```python
df["defect_count_num"] = pd.to_numeric(
    df["defect_count_clean"],
    errors="coerce",
)
df.head()
```

```python
df["work_minutes_clean"] = (
    df["work_minutes"]
    .astype("string")
    .str.strip()
)
df.head()
```

```python
df["work_minutes_num"] = pd.to_numeric(
    df["work_minutes_clean"],
    errors="coerce",
)
df.head()
```

```python
df["occurrence_date_dt"].dtypes
```

```python
df_date_ng = df[
    df["occurrence_date_dt"].isna()
].copy()
df_date_ng
```

```python
#cell 14
df_defect_count_ng = df[
    df["defect_count_num"].isna()
].copy()
df_defect_count_ng
```

```python
#cell 15
df_work_minutes_ng = df[
    df["work_minutes_num"].isna()
].copy()
df_work_minutes_ng
```

```python
df_work = df[
    df["occurrence_date_dt"].notna()
    & df["site_clean"].notna()
    & df["defect_category_clean"].notna()
    & df["defect_count_num"].notna()
    & df["work_minutes_num"].notna()

].copy()
df_work
```

```python
df[
    [
        "defect_count_num",
        "work_minutes_num"
    ]
].dtypes
```

```python
#cell 17
df_work["defect_count_num"] = df_work["defect_count_num"].astype(int)
df_work["work_minutes_num"] = df_work["work_minutes_num"].astype(int)
```

```python
df[
    [
        "defect_count_num",
        "work_minutes_num"
    ]
].dtypes
```

```python
df_work["occurrence_date_day"] = df_work["occurrence_date_dt"].dt.normalize()
df_work
```

```python
df_week = df_work.set_index("occurrence_date_day")
df_week = df_week.sort_index()
df_week
```

```python
weekly_summary = (
    df_week
    .resample("W")
    .agg(
        defect_min=("defect_count_num", "min"),
        defect_max=("defect_count_num", "max"),
        defect_meen=("defect_count_num", "mean"),
        defect_sum=("defect_count_num", "sum"),
        record_count=("defect_count_num", "count"),
        work_minutes_sum=("work_minutes_num", "sum"),
        work_minutes_mean=("work_minutes_num", "mean"),
    )
)
weekly_summary.index.name = "date"
weekly_summary
```

```python
graph_df = weekly_summary.copy()
graph_df.index = graph_df.index.strftime("%Y-%m-%d")
graph_df
```

```python
plt.figure(figsize=(9, 4))

plt.bar(
    graph_df.index,
    graph_df["defect_sum"],
)

plt.title("週ごとの不具合件数")
plt.xlabel("週")
plt.ylabel("不具合件数")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
weekly_site_summary = (
    df_week
    .groupby(
        [
            pd.Grouper(freq="W"),
            "site_clean",
        ]
    )
    .agg(
        defect_sum=("defect_count_num", "sum"),
    )
    .reset_index()
)
weekly_site_summary
```

```python
weekly_site_matrix = weekly_site_summary.pivot(
    index="occurrence_date_day",
    columns="site_clean",
    values="defect_sum",
)

weekly_site_matrix = weekly_site_matrix.fillna(0).astype(int)

weekly_site_matrix.index.name = "date"

weekly_site_matrix
```
