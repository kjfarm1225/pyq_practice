```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = "Meiryo"
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_csv("dataset/study_log_2months.csv",
                 parse_dates=["start_at", "end_at"],
)
df
```

```python
df.head(1)
```

```python
iso_df = df["start_at"].dt.isocalendar()
```

```python
iso_df
```

```python
df["week_group"] = (iso_df["year"] * 100 + iso_df["week"]).astype(int)
df
```

```python
df.head()
```

```python
df["study_minutes"] = (
    df["end_at"] - df["start_at"]
).dt.total_seconds() / 60

df
```

```python
df["study_minutes"]
```

```python
df_duration_ng = df[
    df["study_minutes"] <= 0
].copy()

df_duration_ng
```

```python
df_work = df[
    df["study_minutes"] > 0
].copy()

df_work
```

```python
df_work.head()
```

```python
df_week_summary = df_work.groupby(
    "week_group",
    as_index=False,
).agg(
    study_count=("id", "count"),
    total_minutes=("study_minutes", "sum"),
    avg_minutes=("study_minutes", "mean"),
)
df_week_summary
```

```python
df_week_summary["total_minutes"] = df_week_summary["total_minutes"].round(1)
df_week_summary["avg_minutes"] = df_week_summary["avg_minutes"].round(1)
df_week_summary
```

```python
plt.figure(figsize=(10, 4))
```

```python
plt.bar(
    df_week_summary["week_group"].astype(str),
    df_week_summary["total_minutes"],
)

plt.title("週別学習時間合計")
plt.xlabel("週グループ")
plt.ylabel("学習時間合計")
plt.grid(axis="y")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
