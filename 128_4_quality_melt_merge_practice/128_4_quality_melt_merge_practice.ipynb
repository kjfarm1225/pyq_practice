```python
import pandas as pd
```

```python
index = pd.Index(
    [
        "LCM-A",
        "LCM-B",
        "LCM-C",
        "LCM-D",
        "LCM-E",
    ],
    name="product",
)

columns = pd.Index(
    [
        "MELMB",
        "WSE",
        "GSE",
    ],
    name="site",
)

data = [
    [18, 25, 7],
    [12, 32, 9],
    [5, 14, 21],
    [8, 11, 4],
    [22, 6, 13],
]

df_wide = pd.DataFrame(
    data,
    index=index,
    columns=columns,
)

df_wide
```

```python
df_product_master = pd.DataFrame(
    [
        ["LCM-A", "ヘッドランプ", "品質保証1課"],
        ["LCM-B", "ヘッドランプ", "品質保証1課"],
        ["LCM-C", "リアランプ", "品質保証2課"],
        ["LCM-D", "ヘッドランプ", "品質保証1課"],
        ["LCM-E", "リアランプ", "品質保証2課"],
    ],
    columns=[
        "product",
        "category",
        "owner_department",
    ],
)

df_product_master
```

```python
df_long = df_wide.melt(
    value_name="defect_qty",
    ignore_index=False,
)
print(df_long)
df_long = df_long.reset_index()
df_long
```

```python
df_merge = df_long.merge(
    df_product_master,
    on="product",
)

df_merge.head()
```

```python
df_sort = df_merge.sort_values(
    "defect_qty",
    ascending=False,
)
df_sort
```

```python
df_sort["check_level"] = "通常確認"
df_sort.head()

df_sort.loc[
    df_sort["defect_qty"] >= 20,
    "check_level",
] = "重点確認"
df_sort.head()
```

```python
df_priority = df_sort[
    df_sort["check_level"] == "重点確認"
]
df_priority
```

```python
df_report = df_priority[
    [
        "owner_department",
        "product",
        "site",
        "defect_qty",
        "check_level",
    ]
]
df_report = df_report.reset_index(drop=True)
df_report
```
