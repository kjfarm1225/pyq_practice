###### 126-2: 最小と最大を抽出するには

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    [
        ["Alice", "", 87],
        ["Alice", "", np.nan],
        ["Bob", "", 65],
        ["Bob", "", 92],
    ],
    columns= ["Name", "Subject", "Point"],
)
print(df.Point.min())
print(df.Point.max())
df.Point == df.Point.min()
df.Point == df.Point.max()
#df[(df.Point == df.Point.min()) | (df.Point == df.Point.max())]
```
