###### 126-2: 最小と最大を抽出するには

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    [
        ["Alice", "", 87],
        ["Alice", "", np.nan],
        ["Bob", "", 65],
        ["Bob", "", 92],
    ],
    columns= ["Name", "Subject", "Point"],
)
print(df.Point.min())
print(df.Point.max())
df.Point == df.Point.min()
df.Point == df.Point.max()
#df[(df.Point == df.Point.min()) | (df.Point == df.Point.max())]
```

```python
import pandas as pd

df_quality = pd.DataFrame(
    [
        ["melmb", "a001", 1200.0],
        ["wse", "a002", 850.0],
        ["gse", "a003", 2300.0],
        ["ise", "a004", 400.0],
    ],
    columns=["site", "product", "ppm"],
)

df_quality[
    (df_quality["ppm"] == df_quality["ppm"].min()) |
    (df_quality["ppm"] == df_quality["ppm"].max())
]
```

```python
import pandas as pd

df_defect = pd.DataFrame(
    [
        ["melmb", "lamp-a", 12],
        ["wse", "lamp-b", 5],
        ["gse", "lamp-c", 28],
        ["ise", "lamp-d", 5],
    ],
    columns=["site", "product", "defects"]
)
df_defect[
    (df_defect["defects"] == df_defect["defects"].min()) |
    (df_defect["defects"] == df_defect["defects"].max())
]
```
