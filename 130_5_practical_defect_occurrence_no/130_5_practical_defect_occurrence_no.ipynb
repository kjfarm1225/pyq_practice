```python
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.DataFrame(
    [
        ["2026-06-03", "SUS", "J69P", "D-004", "コイル", 1],
        ["2026-06-01", "SUS", "J69P", "D-001", "NTF", 1],
        ["2026-06-02", "SUS", "310D", "D-002", "Buck-IC", 1],
        ["2026-06-04", "WSE", "310D", "D-005", "コイル", 2],
        ["2026-06-01", "WSE", "310D", "D-003", "NTF", 1],
        ["2026-06-05", "SUS", "J69P", "D-006", "Buck-IC", 1],
        ["2026-06-02", "GSE", "J69P", "D-007", "コイル", 1],
        ["2026-06-03", "GSE", "J69P", "D-008", "CAN通信", 1],
        ["2026-06-06", "WSE", "310D", "D-009", "コイル", 1],
        ["2026-06-02", "SUS", "J69P", "D-010", "コイル", 1],
    ],
    columns=[
        "occ_date",
        "site",
        "model",
        "defect_id",
        "defect_category",
        "defect_qty",
    ],
)
df_defect
```

```python
df_work = df_defect.copy()
df_work
```

```python
df_work = df_work.sort_values(
    [
        "site",
        "model",
        "occ_date",
        "defect_id",
    ]
).reset_index(drop=True)
df_work
```

```python
df_work["occurrence_count0"] = df_work.groupby(
    [
        "site",
        "model",
    ]
).cumcount()
df_work
```

```python
df_work["occurrence_no"] = df_work["occurrence_count0"] + 1
df_work
```

```python
df_result = df_work[
    [
        "site",
        "model",
        "occ_date",
        "defect_id",
        "defect_category",
        "defect_qty",
        #"occurrence_count0",
        "occurrence_no",
    ]
]

df_result
```
