```python
import pandas as pd

names = ["名前", "クラス", "クラブ", "実施月", "数学"]
df = pd.read_excel("input/exam.xlsx", header=1, names=names)

df.head()
```

```python
df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)
df.head()
```

```python
sr1 = df.groupby("名前")["実施月"].count()
sr1.head()
```

```python
result1 = sr1[sr1 < 2].index
result1
```
