```python
import pandas as pd

names = ["名前", "クラス", "クラブ", "実施月", "数学"]
df = pd.read_excel("input/exam.xlsx", header=1, names=names)

df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)

df["クラブ活動"] = ~df["クラブ"].isna()
result = df.groupby("クラブ活動")["数学"].mean()

df[["クラブ", "クラブ活動"]].head()
print(result)
```

```python
#練習1
#クラブ活動ごとの 数学の最高点 を出してください。
#変数名は result_max にしてください。
result_max = df.groupby("クラブ活動")["数学"].max()
result_max
```

```python
#練習2
#クラブ活動ごとの 数学の最低点 を出してください。
#変数名は result_min にしてください。
result_min = df.groupby("クラブ活動")["数学"].min()
result_min
```

```python
#練習3
#クラブ活動ごとの 人数（行数） を出してください。
#変数名は result_count にしてください。
result_count = df.groupby("クラブ活動")["数学"].count()
result_count
```
