```python
import pandas as pd
import numpy as np
```

```python
rnd = np.random.default_rng(3)

df = pd.DataFrame(
    rnd.standard_normal(100),
    columns=[
        "Value",
    ],
)
df
```

```python
labels = [
    "第1四分位数以下",
    "第2四分位数以下",
    "第3四分位数以下",
    "その他",
]
```

```python
df["Category"] = pd.qcut(
    df["Value"],
    4,
    labels=labels,
)
df
```

```python
df_category_count = df["Category"].value_counts(
    sort=False
).reset_index()

df_category_count

df_category_count.columns = [
    "Category",
    "count",
]
df_category_count
```
