```python
import pandas as pd

#step2で作成したcsvを読み込む
df = pd.read_csv(
    "dataset/calories.csv",
    parse_dates=["日時"]
)
df.head()
```

```python
# resampleを使うため、日時をインデックスにする
df = df.set_index("日時")
df.head()
```

```python
summary_df = df.resample("D")[["カロリー", "スキップ"]].sum()
summary_df.head()
```

```python
summary_df = summary_df.rename(
    columns= {"カロリー": "合計カロリー", "スキップ": "スキップ件数"}
)
summary_df.head()
```

```python
names = df["種別"].unique()
names
```

```python
for category in names:
    selected_df= df[df["種別"] == category]
    summary_df[category] = selected_df.resample("D")["カロリー"].sum()
summary_df.head()
```

```python
def is_ok(sr):
    result_morning = sr["朝食"] >= 400
    result_lunch = sr["昼食"] >= 400
    result_dinner = sr["夕食"] >= 400
    result_all = (1500 <= sr["合計カロリー"]) & (sr["合計カロリー"] <= 2100)

    result_sr = (sr["スキップ件数"] >= 1) | (
        result_morning & result_lunch & result_dinner & result_all
    )
    return result_sr
```

```python
summary_df["デイリー目標OK"] = summary_df.apply(is_ok, axis=1)
summary_df.head()
```

```python
summary_df.index.name = "日付"
summary_df.head()
```
