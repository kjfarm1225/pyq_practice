```python
import pandas as pd
df = pd.read_csv("dataset/calories.csv", parse_dates=["日付"]
)
df.head()
```

```python
df[df["カロリー"].isna()]
```

```python
df[df["カロリー"] < 5]
```

```python
df = df[~(df["カロリー"] < 5)]
df
```

```python
mean_lunch = df.loc[
    (df["種別"] == "昼食") & (df["カロリー"] > 1),
    "カロリー"
].mean()
mean_lunch
```

```python
df.loc[
    (df["種別"] == "昼食") & (df["カロリー"].isna()),
    "カロリー"
] = mean_lunch
df.head()
```

```python
df = df.set_index("日付")
df.head()
```

```python
daily_total = df["カロリー"].resample("D").sum().to_frame()
daily_total.head()
```

```python
meal_types = df["種別"].unique()
meal_types
```

```python
for meal in meal_types:
    daily_total[meal] = df.loc[df["種別"] == meal, "カロリー"].resample("D").sum()
daily_total.head()
```

```python
daily_total = daily_total.fillna(0)
daily_total
```

```python
daily_total = daily_total.rename(columns={"カロリー": "合計カロリー"})
daily_total
```

```python
weekly_total = df["カロリー"].resample("W").sum().to_frame()
weekly_total
```

```python
mean_dinner = df.loc[
    (df["種別"] == "夕食") & (df["カロリー"] > 1),
    "カロリー"
].mean()
mean_dinner
```

```python
selected_meals = ["朝食", "夕食"]
daily_small = df["カロリー"].resample("D").sum().to_frame()

for meal in selected_meals:
    daily_small[meal] = df.loc[df["種別"] == meal, "カロリー"].resample("D").sum()

daily_small = daily_small.fillna(0)
daily_small
```
