```python
import pandas as pd
df = pd.read_csv("dataset/calories.csv", parse_dates=["日時"])
df.head()
```

```python
df_select_na = df[df["カロリー"].isna()]
df_select_na
```

```python
df_select_low = df[df["カロリー"] < 5]
df_select_low
```

```python
df= df[~(df["カロリー"].isna() & (~df["スキップ"] == False))]
df
```

```python
mean_lunch = df[(df["種別"] == "昼食") & (df["カロリー"] > 1)]["カロリー"].mean()
mean_lunch
```

```python
df.loc[df["カロリー"] == 1, "カロリー"] = mean_lunch
df
```
