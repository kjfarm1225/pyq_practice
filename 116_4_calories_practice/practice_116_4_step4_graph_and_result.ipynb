```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Meiryo"

summary_df = pd.read_csv(
    "dataset/ex04_summary_df.csv",
    index_col="日付",
    parse_dates=["日付"]
)
summary_df.index = summary_df.index.strftime("%Y-%m-%d")

summary_df.head()
```

```python
category_names = ["朝食", "昼食", "間食", "夕食"]
category_names
```

```python
summary_df[category_names].plot.bar(
    title="日別の摂取カロリー",
    stacked=True
)
```
