## 1問目 ヒストグラム ── データのばらつきを見る

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
""
```

```python
scores_a = [19, 21, 30, 42, 55, 58, 61, 62, 68, 68,
            69, 70, 72, 74, 75, 75, 90, 92, 98, 100]

# クラスBの生徒の点数（20人分）
scores_b = [8, 15, 21, 28, 35, 50, 55, 58, 58, 58,
            58, 65, 66, 67, 70, 71, 72, 73, 86, 95]
```

```python
fig, ax = plt.subplots()

ax.hist(scores_a,
        bins=5,
        range=(0, 100))

fig.suptitle("試験結果の分布（クラスA）")
ax.set_xlabel("試験の点数(点)")
ax.set_ylabel("人数（人）")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.hist((scores_a, scores_b),
        bins=5,
        range=(0, 100),
        label=["クラスA", "クラスB"])

ax.set_title("試験結果の分布")
ax.set_xlabel("試験の点数(点)")
ax.set_ylabel("人数（人）")
ax.legend()

plt.show()
```

```python
fig, ax = plt.subplots()

ax.hist((scores_a, scores_b),
        bins=5,
        range=(0, 100),
        label=["クラスA", "クラスB"],
        stacked=True)
plt.show()
```

## 実務練習1 ヒストグラム ── 測定電圧のばらつきを見る

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] ="Meiryo"
```

```python
# 製品20個分の測定電圧（V）
voltages = [11.8, 12.0, 12.1, 11.9, 12.2,
            12.0, 11.7, 12.1, 12.3, 11.9,
            12.0, 12.2, 11.8, 12.1, 12.0,
            11.9, 12.2, 12.1, 11.8, 12.0]
```

```python
fig, ax = plt.subplots()

ax.hist(voltages,
        bins=6,
        range=(11.5, 12.5),
        edgecolor="black")

ax.grid(True, linestyle="--")
ax.set_axisbelow(True)

fig.suptitle("測定電圧の分布")
ax.set_xlabel("電圧(v)")
ax.set_ylabel("製品数（個）")

plt.show()
```

## 2問目 箱ひげ図 ── 分布を比較する

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
# クラスAの点数（20人分）
scores_a = [19, 21, 30, 42, 55, 58, 61, 62, 68, 68,
            69, 70, 72, 74, 75, 75, 90, 92, 98, 100]

# クラスBの点数（20人分）
scores_b = [8, 15, 21, 28, 35, 50, 55, 58, 58, 58,
            58, 65, 66, 67, 70, 71, 72, 73, 86, 95]
```

```python
fig, ax = plt.subplots()

ax.boxplot(scores_a,
           tick_labels=["クラスA"],
           showmeans=True)

fig.suptitle("クラスAの試験結果の分布")
ax.set_ylabel("点数(点)")
plt.show()
```

```python
fig, ax = plt.subplots()

ax.boxplot((scores_a, scores_b),
            showmeans=True,
            tick_labels=["クラスA", "クラスB"])
```
