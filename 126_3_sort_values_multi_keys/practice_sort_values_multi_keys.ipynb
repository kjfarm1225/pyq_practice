###### 実務寄りの練習問題

```python
import pandas as pd
import numpy as np

df_quality = pd.DataFrame(
    [
        ["2026-06-01", "MELMB", "Lamp-A", "点灯不良", 12000, 18],
        ["2026-06-01", "MELMB", "Lamp-B", "外観キズ", 8500, 4],
        ["2026-06-02", "WSE", "Lamp-C", "寸法NG", 10000, 25],
        ["2026-06-02", "WSE", "Lamp-D", "異物混入", 6000, 2],
        ["2026-06-03", "GSE", "Lamp-E", "はんだ不良", 7000, 16],
        ["2026-06-03", "GSE", "Lamp-F", "点灯不良", 9000, np.nan],
        ["2026-06-04", "ISE", "Lamp-G", "外観キズ", 11000, 6],
        ["2026-06-04", "ISE", "Lamp-H", "寸法NG", 5000, 9],
    ],
    columns=["date", "site", "product", "defect_type", "production", "defects"],
)

df_quality["ppm"] = df_quality["defects"] / df_quality["production"] * 1000000
df_quality["ppm"] = df_quality["ppm"].round(1)
print(df_quality)

df_sorted = df_quality.sort_values(
    ["site", "ppm"],
    ascending=[True, False]
)
print("----- df_sorted -----")
print(df_sorted)

df_sorted_defects = df_quality.sort_values(
    ["site", "defects"],
    ascending=[True, False]
)
print("----- df_sorted_defects -----")
print(df_sorted_defects)



```

```python
# 危険ランク追加
df_sorted["危険ランク"] = np.where(
    df_sorted["ppm"] >= 2000,
    "s級",
    np.where(
        df_sorted["ppm"] >= 1000,
        "a級",
        "通常"
     )
)
print("----- 危険ランク追加 -----")
print(df_sorted)
```
