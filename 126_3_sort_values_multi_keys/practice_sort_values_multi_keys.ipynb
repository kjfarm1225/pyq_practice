###### 126-3: 列ごとに昇順／降順を変えて確認するには

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    [
        ["Alice", "国語", 87],
        ["Alice", "数学", np.nan],
        ["Bob", "国語", 65],
        ["Bob", "数学", 92],
    ],
    columns=["Name", "Subject", "Point"],
)

df.sort_values(["Name", "Point"], ascending=[True, False])
```

```python
import pandas as pd
import numpy as np

df_quality = pd.DataFrame(
    [
        ["melmb", "lamp-a", 12000, 18],
        ["melmb", "lamp-b", 8500, 4],
        ["wse", "lamp-c", 10000, 25],
        ["wse", "lamp-d", 6000, 2],
        ["gse", "lamp-e", 7000, np.nan],
    ],
    columns=["site", "product", "production", "defects"],
)

df_quality["ppm"] = df_quality["defects"] / df_quality["production"] * 1000000
df_quality["ppm"] = df_quality["ppm"].round(1)

df_quality
```

```python
df_quality.sort_values(["site", "ppm"], ascending=[True, False])
```

```python
df_quality.sort_values(["site", "defects"], ascending=[True, False])
```
