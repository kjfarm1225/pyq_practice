## 「136-1: NumPデータの連結・分割」 1問目

```python
import numpy as np

rainfalls_ab = np.array([[120, 130, 140],
                        [115, 140, 135]])

rainfalls_cd = np.array([[130, 125, 140],
                        [90, 120, 100]])

rainfalls_ab_2 = np.array([[90, 100, 115],
                          [120, 95, 105]])
```

```python
rainfalls_ab_cd = np.vstack([rainfalls_ab, rainfalls_cd])
rainfalls_ab_cd
```

```python
rainfalls_ab_all = np.hstack([rainfalls_ab, rainfalls_ab_2])
rainfalls_ab_all
```

```python
np.concatenate([rainfalls_ab, rainfalls_cd], axis=0)
```

```python
np.concatenate([rainfalls_ab, rainfalls_ab_2], axis=1)
```
