```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
```

```python
names = ["名前", "クラス", "クラブ", "実施月", "数学"]

df = pd.read_excel("input/exam.xlsx", header=1, names=names)
df[:4]
```

```python
df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)
df[:4]
```

```python
for k, sr in df.groupby("クラス")["数学"]:
    sr.plot.kde(label=k)
plt.legend();
```

```python
#練習1
#クラス別ではなく、実施月ごと に数学の KDE グラフを描いてください

import matplotlib.pyplot as plt

for k, sr in df.groupby("実施月")["数学"]:
    sr.plot.kde(label=k)
plt.legend();
```
