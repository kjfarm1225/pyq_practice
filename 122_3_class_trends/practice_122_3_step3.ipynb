```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
```

```python
names = ["名前", "クラス", "クラブ", "実施月", "数学"]

df = pd.read_excel("input/exam.xlsx", header=1, names=names)

df[:4]
```

```python
df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)
df[:4]
```

```python
df.groupby("実施月")["数学"].mean()
```

```python
for k, sr in df.groupby("実施月")["数学"]:
    sr.hist(range=(0, 100), bins=20, label=k, alpha=0.5)
plt.legend();
```

```python
#練習1
#棒の数を20ではなく 10 に変えてください。
for k, sr in df.groupby("実施月")["数学"]:
    sr.hist(range=(0, 100), bins=10, label=k, alpha=0.5)
```

```python
#練習2
#横軸の範囲を 30〜100点 に変えてください。
for k, sr in df.groupby("実施月")["数学"]:
    sr.hist(range=(30, 100), bins=20, label=k, alpha=0.5)
plt.legend();
```

```python
#練習3
#ヒストグラムではなく、KDEグラフに変えて表示してください。
for k, sr in df.groupby("クラス")["数学"]:
    sr.plot.kde(label=k)
plt.legend();
```
