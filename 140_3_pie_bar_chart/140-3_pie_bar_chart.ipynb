## 1問目 円グラフ ── 構成比を見る

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
labels = ["満足", "普通", "不満", "未回答"]
counts = [132, 89, 20, 5]
```

```python
fig, ax = plt.subplots()

ax.pie(counts, labels=labels,
        startangle=90,
        counterclock=False,
        autopct="%1.1f%%")

ax.set_title("アンケート結果の構成比")
plt.show()
```
