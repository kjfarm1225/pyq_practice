# 140-3 円グラフ・棒グラフ
## 3問目 積み上げ棒グラフ ── カテゴリーごとに数量を比較する

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
labels = ["満足", "普通", "不満", "無回答"]
counts_a = [132, 89, 20, 5]
counts_b = [72, 59, 5, 1]
```

```python
fig, ax = plt.subplots()

ax.bar([1, 2, 3, 4],
       counts_a,
       tick_label=labels,
       label="A会場")

ax.bar([1, 2, 3, 4],
       counts_b,
       bottom=counts_a,
       tick_label=labels,
       label="B会場")

ax.set_title("アンケート結果の人数")
ax.set_xlabel("回答")
ax.set_ylabel("人数")
ax.legend()

plt.show()
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
months = ["1月", "2月", "3月", "4月"]
parts_defect = [4, 7, 5, 6]
assembly_defect = [2, 3, 1, 8]
```

```python
fig, ax = plt.subplots()
ax.bar(months,
       parts_defect,
       label="部品不良")

ax.bar(months,
       assembly_defect,
       bottom=parts_defect,
       label="組立不良")

ax.set_title("月別工程別不具合件数")
ax.set_xlabel("発生月")
ax.set_ylabel("不具合件数（件）")
ax.legend()

plt.show()
```

```python
# (1) 描画オブジェクト（fig）とサブプロット（ax）を作成
fig, ax = plt.subplots()

# (2) サブプロットに対して、グラフを描画
x_list = [1, 2, 3, 4]  # X座標のデータ（位置）
width = 0.3 # 棒の太さ

# ▼ポイント1: 1つ目の棒グラフをやや細めにして描画
ax.bar(x_list,  # X座標のデータ（位置）
       counts_a,
       tick_label=labels,
       label="A会場",
       width=width)  # 棒の太さをやや細めに設定

# ▼ポイント2: 2つ目の棒グラフを右にずらして描画
ax.bar([x + width for x in x_list],  # 1つ目のグラフから棒の太さだけ右にずらす
       counts_b,
       label="B会場",
       width=width)  # 棒の太さをやや細めに（1つ目のグラフと同じ）

# タイトルとラベルの設定
ax.set_title("アンケート結果の人数")
ax.set_xlabel("回答")
ax.set_ylabel("人数（人）")
ax.legend()

# (3) 図全体を表示
plt.show()
```
