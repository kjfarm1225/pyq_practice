## 135-3: NumPyで乱数を作ろう

```python
import numpy as np
```

```python
rng = np.random.default_rng()
```

```python
rnd1 = rng.normal(50, 10, 10)
rnd1
```

```python
rnd2 = rng.normal(50, 10, (2, 5))
rnd2
```

```python
rnd3 = rng.standard_normal(10)
rnd3
```

```python
rnd4 = rng.binomial(20, 0.4, 10)
rnd4
```

```python
rnd5 = rng.poisson(9.8, 10)
rnd5
```

```python
rnd6 = rng.beta(1, 2, 10)
rnd6
```

```python
rnd7 = rng.multivariate_normal([0, 0], [[1, 0.5], [0.5, 1]])
rnd7
```

## 問題2

成功確率50%の試行を100回行ったときの成功回数を、10個作り、変数 change2 に入れてください。

```python
change2 = rng.binomial(100, 0.5, 10)
change2
```
