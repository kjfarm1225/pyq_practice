```python
#cell 1
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        ["2026-06-01", "08:00", "J69P", "MELMB", "Line1", "L250601-A", "コイル", 300, 0],
        ["2026-06-01", "10:00", "", "", "", "", "コイル", 280, 1],
        ["2026-06-01", "13:00", "", "", "", "", "Buck-IC", 320, 2],
        ["2026-06-02", "08:00", "310D", "WSE", "Line2", "L250602-B", "はんだ", 250, 0],
        ["2026-06-02", "10:00", "", "", "", "", "コイル", 260, 1],
        ["2026-06-02", "13:00", "", "", "", "", "NTF", 240, 0],
        ["2026-06-03", "08:00", "J69P", "GSE", "Line1", "L250603-C", "コイル", 310, 3],
        ["2026-06-03", "10:00", "", "", "", "", "Buck-IC", 290, 0],
        ["2026-06-03", "13:00", "", "", "", "", "コイル", 300, 1],
        ["2026-06-04", "08:00", "310D", "MELMB", "Line3", "L250604-D", "はんだ", 0, 1],
        ["2026-06-04", "10:00", "", "", "", "", "コイル", 270, 2],
        ["2026-06-04", "13:00", "", "", "", "", "NTF", 260, 0],
    ],
    columns=[
        "inspection_date",
        "inspection_time",
        "project",
        "site",
        "line",
        "lot_no",
        "defect_category",
        "inspection_qty",
        "defect_qty",
    ],
)
df
```

```python
#cell 3
df_work = df.copy()
df_work
```

```python
df_work = df_work.replace(
    "",
    np.nan,
)
df_work
```

```python
#cell 4
df_work["project_filled"] = df_work["project"].ffill()
df_work["site_filled"] = df_work["site"].ffill()
df_work["lot_no_filled"] = df_work["lot_no"].ffill()
df_work["line_filled"] = df_work["line"].ffill()

df_work
```

```python
#cell 5

df_clean = df_work[
    [
        "inspection_date",
        "inspection_time",
        "project_filled",
        "site_filled",
        "line_filled",
        "lot_no_filled",
        "defect_category",
        "inspection_qty",
        "defect_qty",
    ]
].copy()
df_clean
```

```python
df_clean["defect_rate_safe"] = (
    (df_clean["defect_qty"] / df_clean["inspection_qty"] * 1_000_000)
    .where(df_clean["inspection_qty"] > 0, 0)
    .round(1)
)

df_clean
```

```python
#cell 7
df_clean["check_result"] = "通常"

df_clean.loc[
    df_clean["defect_rate_safe"] > 5000,
    "check_result",
 ] = "要確認"
df_clean
```

```python
df_check = df_clean[
    df_clean["check_result"] == "要確認"
].copy()

df_check
```

```python
df_check_sort = df_check.sort_values(
    "defect_rate_safe",
    ascending=False
)
df_check_sort
```

```python
df_cross = df_clean.pivot_table(
    values="defect_qty",
    index="line_filled",
    columns="defect_category",
    aggfunc="sum",
    fill_value=0,
)
df_cross
```

```python
df_cross["total"] = df_cross.sum(axis=1)
df_cross
```

```python
df_cross_sort = df_cross.sort_values(
    "total",
    ascending=False
)
df_cross_sort
```
