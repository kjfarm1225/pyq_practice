```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("dataset/room.csv")
```

```python
pattern = r"^\d号棟\d階\d{3}号室$"
```

```python
selected_df = df[
    df["room"].str.match(
        pattern,
        na=False,
    )
].copy()
selected_df
```
