```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")
plt.rcParams["font.family"] = "Meiryo"
```

```python
mu = 1.20
sigma = 0.04
```

```python
lower_limit = 1.05
upper_limit = 1.35
```

```python
rng = np.random.default_rng(7)
```

```python
normal_torque = rng.normal(mu, sigma, 970)
weak_torque = rng.normal(1.02, 0.02, 30)
```

```python
torque = np.concatenate([normal_torque, weak_torque])
```

```python
# ヒストグラムの描画：ざっくり確認
fig = plt.figure(figsize=(10, 5))
ax = fig.add_subplot(1, 1, 1)

ax.hist(torque, rwidth=0.9, bins=16)

ax.axvline(lower_limit, linestyle="--", color="red", label="下限規格")
ax.axvline(upper_limit, linestyle="--", color="red", label="上限規格")

ax.set_title("品質モンスター分布監査：トルク値ヒストグラム")
ax.set_xlabel("トルク値 N・m")
ax.set_ylabel("件数")
ax.legend();
plt.show()


# ヒストグラムの描画：細かく確認
fig = plt.figure(figsize=(10,5))
ax = fig.add_subplot(1, 1, 1)

ax.hist(torque, rwidth=0.9, bins=40)

ax.axvline(lower_limit, linestyle="--", color="red", label="下限規格")
ax.axvline(upper_limit, linestyle="--", color="red", label="上限規格")

ax.set_title("品質モンスター分布監査：トルク値ヒストグラム bins=40")
ax.set_xlabel("トルク値 N・m")
ax.set_ylabel("件数")
ax.legend();
plt.show()
```
