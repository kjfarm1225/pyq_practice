```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')
plt.rcParams["font.family"] = "Meiryo"
```

```python
rng = np.random.default_rng(12)
```

```python
good_score = rng.normal(7.8, 0.7, 260)
bad_score = rng.normal(5.2, 0.5, 80)
```

```python
quality_score = np.concatenate([good_score, bad_score])
data = quality_score
bins = np.arange(0, 10.5, 0.5)
rwidth = 0.7
```

```python
fig = plt.figure(figsize=(10, 5))
ax = fig.add_subplot(1, 1, 1)
ax.hist(data, bins=bins, rwidth=rwidth)
ax.axvline(6.0, linestyle="--", color="red", label="注意ライン 6.0点")
ax.set_xlim(0, 10)
ax.set_xticks(np.arange(0, 11, 1));

ax.set_title("品質モンスター分布調査：仕入れ先品質スコア")
ax.set_xlabel("品質スコア")
ax.set_ylabel("件数")
ax.legend(loc="upper left");
plt.show()
```
