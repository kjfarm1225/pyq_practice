```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")
plt.rcParams["font.family"] = "Meiryo"
```

```python
mu = 50.0
sigma = 0.8
```

```python
rng = np.random.default_rng(3)
x = rng.normal(mu, sigma, 10000)
```

```python
fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.hist(x, rwidth=0.9, bins=10)

ax.set_title("寸法測定値のヒストグラム")
ax.set_xlabel("寸法値")
ax.set_ylabel("件数")

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.hist(x, rwidth=0.9, bins=30)

ax.set_title("寸法測定値のヒストグラム:bins=30")
ax.set_xlabel("寸法値")
ax.set_ylabel("件数")
```
