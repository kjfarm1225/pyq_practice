## 演習問題

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np

plt.style.use("ggplot")
plt.rcParams["font.family"] = "Meiryo"

mu = 100
sigma = 0.8

rng = np.random.default_rng(1)
x = rng.normal(mu, sigma, 10000)

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)

ax.hist(x)

ax.set_title("品質モンスター調査：寸法測定値の分布")
ax.set_xlabel("寸法測定値")
ax.set_ylabel("件数")
```

## 練習問題２

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")
plt.rcParams["font.family"] = "Meiryo"
```

```python
mu = 50.0
sigma = 20
```

```python
rng = np.random.default_rng(1)
x = rng.normal(mu, sigma, 10000)

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)

ax.hist(x)

ax.set_title("品質モンスター調査：ばらつき大の寸法分布")
ax.set_xlabel("寸法値")
ax.set_ylabel("件数")
```
