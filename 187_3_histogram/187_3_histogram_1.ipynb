```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.style.use("ggplot")
```

```python
mu = 100
sigma = 10
```

```python
rng = np.random.default_rng(0)
x = rng.normal(mu, sigma, 10000)
```

```python
fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.hist(x);
```
