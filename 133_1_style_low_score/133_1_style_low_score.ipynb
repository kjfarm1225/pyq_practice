```python
import pandas as pd
```

```python
df = pd.read_csv(
    "dataset/score.csv",
)
df
```

```python
def style_low_score(x):
    style = "font-weight: bold;" if x < 70 else None
    return style
```

```python
df.style.map(
    style_low_score,
    subset=["国語", "数学", "理科", "社会"],
)
```
