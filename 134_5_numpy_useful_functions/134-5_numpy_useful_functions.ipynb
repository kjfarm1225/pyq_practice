```python
import numpy as np

defects = np.array([5, 7, 4, 4, 9, 6])

mean_defects = defects.mean()
max_defects = defects.max()
diff_defects = np.diff(defects)
sum_defects = defects.sum()

mean_defects, max_defects, diff_defects, sum_defects
```

```python
import numpy as np

temps = np.array([20, 22, 25, 28, 27, 30, 24])
temps
```

```python
diff_temps = np.diff(temps)
diff_temps
```

```python
abs_diff_temps = np.abs(diff_temps)
abs_diff_temps
```

```python
np.abs(diff_temps)
```

```python
result = []

for x in diff_temps:
    result.append(abs(x))
```

```python
result
```

```python
np.abs(diff_temps)
```
