```python
import pandas as pd
df = pd.read_csv("dataset/reading.txt", sep=" ", encoding="utf-8-sig")
df.head()
```

```python
for char in ["-", "年", "組"]:
    df["クラス"] = df["クラス"].str.replace(char, "", regex=False)

df.head()
```

```python
valid_title = "キューちゃん海へゆく"
df.loc[df["タイトル"].str.contains("キューちゃん"), "タイトル"] = valid_title
df["タイトル"].unique()
```

```python
df["学年"] = df["クラス"].str[0] + "年"
df["組"] = df["クラス"].str[1] + "組"
df.head()
```

```python
df.to_csv("dataset/reading_after.txt", sep=" ", index=False, encoding="utf-8-sig")
```

```python
user_grouped = df.groupby(["学年", "組", "出席番号"])

user_df = user_grouped.agg({
    "タイトル": "nunique",
    "ページ数": "sum"
}).reset_index()

user_df.head()
```

```python
class_df = user_df.groupby(["学年", "組"]).agg({
    "タイトル": "sum",
    "ページ数": "mean",
    "出席番号": "nunique"
})

class_df = class_df.rename(columns={
    "タイトル": "読んだ本の数",
    "ページ数": "平均ページ数",
    "出席番号": "参加人数",
})
class_df
```

```python
rank_df = class_df.rank(ascending=False)
```

```python
rank_df = rank_df.add_suffix("_全体順位")
```

```python
class_df[rank_df.columns] = rank_df
```

```python
class_df
```

```python
class_df.to_csv("dataset/class_df.txt", sep=" ", encoding="utf-8-sig")
```

```python
class_df = class_df.reset_index()
```

```python
grouped = class_df.groupby("学年")
rank_df = grouped[["読んだ本の数", "平均ページ数", "参加人数"]].rank(ascending=False)
rank_df = rank_df.add_suffix("_学年順位")
class_df[rank_df.columns] = rank_df
class_df
```

```python
class_df.to_csv("dataset/reading_ranking.txt", sep=" ", index=False, encoding="utf-8-sig")
```
