```python
import pandas as pd

df = pd.read_csv("dataset/reading_after.txt", sep=" ", encoding="utf-8-sig")
df.head()
```

```python
df.columns
```

```python
user_grouped = df.groupby(["学年", "組", "出席番号"])
user_grouped
```

```python
user_df = user_grouped.agg({

    "タイトル": "nunique",
    "ページ数": "sum"
})
user_df
```

```python
user_df = user_df.reset_index()
user_df
```

```python
grouped = user_df.groupby(["学年", "組"])
grouped
```

```python
class_df = grouped.agg({
    "タイトル":"sum",
    "ページ数":"mean",
    "出席番号":"nunique"
})
class_df
```

```python
class_df = class_df.rename(columns={
    "タイトル": "読んだ本の数",
    "ページ数": "平均ページ数",
    "出席番号": "参加人数",
})
class_df
```

```python
rank_df = class_df.rank(ascending=False)
rank_df
```

```python
rank_df = rank_df.add_suffix("_全体順位")
rank_df
```

```python
class_df[rank_df.columns] = rank_df
class_df
```

```python
user_grouped = df.groupby(["学年","組", "出席番号"])

user_df = user_grouped.agg({
    "タイトル": "nunique",
    "ページ数": "sum"
})

user_df = user_df.reset_index()
user_df.head()
```

```python
grouped = user_df.groupby(["学年", "組"])

class_df = grouped.agg({
    "タイトル": "sum",
    "ページ数": "mean",
    "出席番号": "nunique"
})

class_df = class_df.rename(
    columns={
        "タイトル": "読書数",
        "ページ数": "平均ページ数",
        "出席番号": "生徒数"
    }
)
class_df.head()
```
