```python
import pandas as pd
from IPython.display import display

def query_bot(fnam, max_rows=5, **option):
    df = pd.read_excel(fnam, **option)
    df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)
    
    with pd.option_context("display.max_rows", max_rows):
        while s := input("> "):
            try:
                if s.isdigit():
                    i = int(s)
                    display(df[i:i + 1])
                elif s == "columns":
                    print(df.columns)
                else:
                    result = df.query(s)
                    display(result)
            except:
                    print("クエリを正しくしてください")
names = ["名前", "クラス", "クラブ", "実施月", "数学"]
query_bot("input/exam.xlsx", header=1, names=names)
```

```python
#練習2
#数字を入力したとき、行だけでなく
#"3行目を表示します" のようなメッセージも出してください。
import pandas as pd
from IPython.display import display

def query_bot(fnam, max_rows=5, **option):
    df = pd.read_excel(fnam, **option)
    df["名前"] = df["名前"].str.replace("u\3000", " ", regex=False)

    with pd.option_context("display.max_rows", max_rows):
        while s := input("> "):
            try:
                if s.isdigit():
                    i = int(s)
                    print(f"{i}行目を表示します")
                    display(df[i:i +1])
                elif s == "columns":
                    print(df.columns)
                else:
                    result = df.query(s)
                    display(result)
            except:
                    print("クエリを正しくしてください")
names = ["名前", "クラス", "クラブ", "実施月", "数学"]
query_bot("input/exam.xlsx", header=1, names=names)
```

```python
#練習3
#shape と入力したら、行数・列数を表示する機能を追加してください。
import pandas as pd
from IPython.display import display

def query_bot(fnam, max_rows=5, **option):
    df = pd.read_excel(fnam, **option)
    df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)

    with pd.option_context("display.max_rows", max_rows):
        while s := input("> "):
            try:
                if s.isdigit():
                    i =int(s)
                    display(df[i:i +1])
                elif s == "columns":
                    print(df.columns)
                elif s == "shape":
                    print(df.shape)
                else:
                    result = df.query(s)
                    display(result)
            except:
                print("クエリを正しくしてください")
names = ["名前", "クラス", "クラブ", "実施月", "数学"]
query_bot("input/exam.xlsx", header=1, names= names)
```
