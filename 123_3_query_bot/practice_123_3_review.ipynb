```python
import pandas as pd
from IPython.display import display

df = pd.DataFrame({
    "名前": ["田中", "佐藤", "鈴木"],
    "点数": [80, 90, 70]
})

def row_bot():
    while s := input("> "):
        if s.isdigit():
            i = int(s)
            display(df[i:i + 1])
        elif s == "columns":
            print(df.columns)
        else:
            print("数字か columns を入れてください")

row_bot()
```
