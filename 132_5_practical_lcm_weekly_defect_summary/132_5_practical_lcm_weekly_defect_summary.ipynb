```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = "Meiryo"
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_csv(
    "dataset/lcm_weekly_defect_practice.csv",
    encoding="utf-8-sig",
)
df.head()
```

```python
df["defect_qty_num"] = pd.to_numeric(
    df["defect_qty"],
    errors="coerce",
)
df.dtypes
```

```python
df["inspection_qty_num"] = pd.to_numeric(
    df["inspection_qty"],
    errors="coerce",
)
df.dtypes
```

```python
df.head()
```

```python
df["occ_at_jst_raw"] = df["occ_at_jst"]
```

```python
df["occ_at_jst_clean"] = (
    df["occ_at_jst"]
    .astype("string")
    .str.strip()
)
df
```

```python
df.head()
```

```python
df.dtypes
```

```python
df["occ_at_jst_dt"] = pd.to_datetime(
    df["occ_at_jst_clean"],
    errors="coerce",
)
df.dtypes
```

```python
df_data_ng = df[
    df["occ_at_jst_dt"].isna()
].copy()
df_data_ng
```

```python
df_qty_ng = df[
    df["defect_qty_num"].isna()
].copy()
df_qty_ng
```

```python
df_inspection_ng = df[
    df["inspection_qty_num"].isna()
].copy()
df_inspection_ng
```

```python
df_work = df[
    df["occ_at_jst_dt"].notna()
    & df["defect_qty_num"].notna()
    & df["inspection_qty_num"].notna()
].copy()
```

```python
df_work
```

```python
df_work.head()
```

```python
iso_df = df_work["occ_at_jst_dt"].dt.isocalendar()
iso_df.head()
```

```python
df_work["week_group"] = (
    iso_df["year"] * 100 + iso_df["week"]
).astype(int)
df_work.head()
```

```python
df_work["ppm"] = (
    df_work["defect_qty_num"]
    / df_work["inspection_qty_num"]
    * 1000000
).round(1)
df_work.head()
```

```python
df_week_summary = df_work.groupby(
    "week_group",
    as_index=False,
).agg(
    defect_count=("defect_id", "count"),
    defect_qty_sum=("defect_qty_num", "sum"),
    inspection_qty_sum=("inspection_qty_num", "sum"),
)
df_week_summary
```

```python
df_week_summary["ppm"] = (
    df_week_summary["defect_qty_sum"]
    / df_week_summary["inspection_qty_sum"]
    * 1000000
).round(1)
```

```python
df_week_summary["defect_qty_sum"] = (
    df_week_summary["defect_qty_sum"].astype(int)
)
```

```python
df_week_summary["inspection_qty_sum"] = (
    df_week_summary["inspection_qty_sum"].astype(int)
)
```

```python
df_week_summary
```

```python
df_work.head()
```

```python
df_week_site_summary = df_work.groupby(
    [
        "week_group",
        "site",
    ],
    as_index=False
).agg(
    defect_count=("defect_id", "count"),
    defect_qty_sum=("defect_qty_num", "sum"),
    inspection_qty_sum=("inspection_qty_num", "sum")
)
df_week_site_summary.head()
```

```python
df_week_site_summary["ppm"] = (
    df_week_site_summary["defect_qty_sum"]
    / df_week_site_summary["inspection_qty_sum"]
    * 1000000
).round(1)
```

```python
df_week_site_summary.head()
```

```python
df_week_site_summary["defect_qty_sum"] = (
    df_week_site_summary["defect_qty_sum"].astype(int)
)
df_week_site_summary.head()
```

```python
df_week_site_summary["inspection_qty_sum"] = (
    df_week_site_summary["inspection_qty_sum"].astype(int)
)
df_week_site_summary
```

```python
df_week_category_summary = df_work.groupby(
    [
        "week_group",
        "defect_category",
    ],
    as_index=False
).agg(
    defect_count=("defect_id", "count"),
    defect_qty_sum=("defect_qty_num", "sum"),
    inspection_qty_sum=("inspection_qty_num", "sum"),
)
df_week_category_summary.head()
```

```python
df_week_category_summary["ppm"] = (
    df_week_category_summary["defect_qty_sum"]
    / df_week_category_summary["inspection_qty_sum"]
    * 1000000
).round(1)
df_week_category_summary["ppm"].head()
```

```python
df_week_category_summary["defect_qty_sum"] = (
    df_week_category_summary["defect_qty_sum"].astype(int)
)
df_week_category_summary.dtypes
```

```python
df_week_category_summary["inspection_qty_sum"] = (
    df_week_category_summary["inspection_qty_sum"].astype(int)
)
df_week_category_summary
```

```python
#cell 19
plt.figure(figsize=(10, 4))
```

```python
plt.plot(
    df_week_summary["week_group"].astype(str),
    df_week_summary["defect_qty_sum"],
    marker="o",
)

plt.title("週別不具合数量推移")
plt.xlabel("週グループ")
plt.ylabel("不具合数量")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
