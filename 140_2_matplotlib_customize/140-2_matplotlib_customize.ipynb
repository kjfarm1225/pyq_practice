# 140-2 Matplotlibのカスタマイズ
## 1問目 複数のグラフを重ねて表示

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps_max = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
y_temps_min = [2.4, 6.2, 7.5, 14.3, 15.7, 20.3]
```

```python
fig, ax = plt.subplots()

ax.plot(x_months, y_temps_max)
ax.plot(x_months, y_temps_min)

fig.suptitle("月別の最高・最低気温")
ax.set_xlabel("月")
ax.set_ylabel("気温（度）")

plt.show()
```
