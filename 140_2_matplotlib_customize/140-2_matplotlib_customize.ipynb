# 140-2 Matplotlibのカスタマイズ
## 1問目 複数のグラフを重ねて表示

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps_max = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
y_temps_min = [2.4, 6.2, 7.5, 14.3, 15.7, 20.3]
```

```python
fig, ax = plt.subplots()

ax.plot(x_months, y_temps_max)
ax.plot(x_months, y_temps_min)

fig.suptitle("月別の最高・最低気温")
ax.set_xlabel("月")
ax.set_ylabel("気温（度）")

plt.show()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

x_months = [1, 2, 3, 4]
y_temps_max = [10, 14, 18, 23]
y_temps_min = [2, 5, 9, 13]

fig, ax = plt.subplots()

ax.plot(x_months, y_temps_max)
ax.plot(x_months, y_temps_min)

fig.suptitle("月別気温")
ax.set_xlabel("月")
ax.set_ylabel("気温")
plt.show()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps_max = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
y_temps_min = [2.4, 6.2, 7.5, 14.3, 15.7, 20.3]
y_temps_avg = [7.4, 10.8, 11.9, 18.6, 21.2, 24.3]
```

```python
fig, ax = plt.subplots()
ax.plot(x_months, y_temps_max)
ax.plot(x_months, y_temps_min)
ax.plot(x_months, y_temps_avg)

fig.suptitle("月別の最高・最低・平均気温")
ax.set_xlabel("月")
ax.set_ylabel("気温（℃）")

plt.show()
```

# 140-2 Matplotlibのカスタマイズ
## 2問目 凡例の表示

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps_max = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
y_temps_min = [2.4, 6.2, 7.5, 14.3, 15.7, 20.3]
```

```python
fig, ax = plt.subplots()
ax.plot(x_months, y_temps_max, label="最高気温")
ax.plot(x_months, y_temps_min, label="最低気温")

fig.suptitle("月別の最高・最低気温")
ax.set_xlabel("月")
ax.set_ylabel("気温（℃）")

ax.legend(loc="lower right")
plt.show()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

x_months = [1, 2, 3, 4]
y_temps_max = [10, 14, 18, 23]
y_temps_min = [2, 5, 9, 13]

fig, ax = plt.subplots()

ax.plot(x_months, y_temps_max, label="最高気温")
ax.plot(x_months, y_temps_min, label="最低気温")

fig.suptitle("4か月の気温")
ax.set_xlabel("月")
ax.set_ylabel("気温")

ax.legend()
plt.show()
```
