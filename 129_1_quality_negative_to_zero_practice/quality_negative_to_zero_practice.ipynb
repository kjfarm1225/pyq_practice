```python
import pandas as pd
```

```python
df = pd.DataFrame(
    [
         ["D001", "J69P", "MELMB", 1200, 3],
        ["D002", "J69P", "WSE", 980, -2],
        ["D003", "310D", "GSE", -500, 1],
        ["D004", "310D", "ISE", 760, 0],
        ["D005", "J69P", "MELMB", 1100, 5],
        ["D006", "310D", "WSE", 850, -1],
        ["D007", "J69P", "GSE", 920, 2],
    ],
    columns=[
        "defect_id",
        "project",
        "site",
        "inspection_qty",
        "defect_qty",
    ],
)

df
```

```python
df["defect_qty_fixed"] = df["defect_qty"].where(
    df["defect_qty"] >= 0,
    0,
)
df
```

```python
#cell4
df["inspection_qty_fixed"] = df["inspection_qty"].where(
    df["inspection_qty"] >= 0,
    0,
)
df
```

```python
#df["defect_rate"] = (df["defect_qty_fixed"] / df["inspection_qty_fixed"] * 1000000).round(1)
#df

df["defect_rate_safe"] = (
    df["defect_qty_fixed"] / df["inspection_qty_fixed"]
    ).where(
    df["inspection_qty_fixed"] > 0,
    0,
)

df
```

```python
#cell6

df_sort = df.sort_values(
    "defect_rate_safe",
    ascending=False,
)
df_sort
```
