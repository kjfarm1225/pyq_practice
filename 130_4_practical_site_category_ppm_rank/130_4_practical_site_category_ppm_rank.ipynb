```python
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.DataFrame(
    [
        ["2026-06", "SUS", "J69P", "コイル", 3, 12000],
        ["2026-06", "SUS", "J69P", "Buck-IC", 1, 12000],
        ["2026-06", "SUS", "310D", "NTF", 2, 8000],
        ["2026-06", "SUS", "310D", "コイル", 0, 8000],
        ["2026-06", "WSE", "J69P", "Buck-IC", 2, 6000],
        ["2026-06", "WSE", "J69P", "コイル", 1, 6000],
        ["2026-06", "WSE", "310D", "コイル", 4, 5000],
        ["2026-06", "WSE", "310D", "NTF", 1, 5000],
        ["2026-06", "GSE", "J69P", "コイル", 2, 4000],
        ["2026-06", "GSE", "J69P", "Buck-IC", 1, 4000],
        ["2026-06", "GSE", "310D", "NTF", 3, 3000],
        ["2026-06", "GSE", "310D", "CAN通信", 1, 3000],
        ["2026-06", "MELMB", "310D", "コイル", 5, 10000],
        ["2026-06", "MELMB", "310D", "NTF", 0, 10000],
        ["2026-06", "MELMB", "J69P", "Buck-IC", 2, 9000],
        ["2026-06", "MELMB", "J69P", "CAN通信", 1, 9000],
    ],
    columns=[
        "month",
        "site",
        "model",
        "defect_category",
        "defect_qty",
        "production_qty",
    ],
)
df_defect
```

```python
df_work = df_defect.copy()
df_work
```

```python
df_summary = df_work.groupby(
    [
        "site",
        "defect_category",
    ],
    as_index=False,
).agg(
    defect_qty_total=("defect_qty", "sum"),
    production_qty_total=("production_qty", "sum"),
)
df_summary
```

```python
df_summary["defect_ppm"] = 0.0
df_summary
```

```python
df_summary.loc[
    df_summary["production_qty_total"] > 0,
    "defect_ppm",
] = (
    df_summary["defect_qty_total"]
    / df_summary["production_qty_total"]
    * 1000000
)
df_summary
```

```python
df_summary["ppm_rank"] = df_summary.groupby("site")["defect_ppm"].rank(
    ascending=False
)
df_summary
```

```python
df_summary["priority"] = "通常確認"
df_summary
```

```python
df_summary.loc[
    df_summary["ppm_rank"] == 1.0,
    "priority",
] = "最優先"
df_summary
```

```python
df_summary.loc[
    df_summary["ppm_rank"] == 2.0,
    "priority",
] = "要確認"

df_summary
```

```python
df_summary_sorted = df_summary.sort_values(
    [
        "site",
        "ppm_rank",
    ]
).reset_index(drop=True)

df_summary_sorted
```
