###### 130_4_practical_site_model_defect_rank

```python
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.DataFrame(
    [
        ["D001", "SUS", "J69P", "コイル", 3],
        ["D002", "SUS", "J69P", "Buck-IC", 1],
        ["D003", "SUS", "310D", "NTF", 1],
        ["D004", "WSE", "310D", "コイル", 4],
        ["D005", "WSE", "310D", "NTF", 2],
        ["D006", "WSE", "J69P", "Buck-IC", 3],
        ["D007", "GSE", "J69P", "コイル", 2],
        ["D008", "GSE", "J69P", "Buck-IC", 3],
        ["D009", "GSE", "310D", "NTF", 2],
        ["D010", "SUS", "310D", "コイル", 0],
    ],
    columns=[
        "defect_id",
        "site",
        "model",
        "defect_category",
        "defect_qty",
    ],
)

df_defect
```

```python
df_work = df_defect.copy()
df_work
```

```python
df_work_positive = df_work[
    df_work["defect_qty"] > 0
].copy()
df_work_positive
```

```python
df_summary = df_work_positive.groupby(
    [
        "site",
        "model",
    ],
    as_index=False,
)["defect_qty"].sum()
df_summary
```

```python
#cell 3
df_summary = df_summary.rename(
    columns={
        "defect_qty": "defect_qty_total",
    }
)
df_summary
```

```python
df_summary["defect_rank"] = df_summary.groupby("site")["defect_qty_total"].rank(
    ascending=False
)
df_summary
```

```python
df_summary_soreted = df_summary.sort_values(
    [
        "site",
        "defect_rank",
    ]
).reset_index(drop=True)
df_summary_soreted
```
