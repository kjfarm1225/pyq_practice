## 少し変更練習問題

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

plt.style.use("ggplot")
```

```python
rng = np.random.default_rng(2)
```

```python
x = np.arange(1, 51)
```

```python
y = 4 * x * rng.random(50)
```

```python
fig = plt.figure()
ax = fig.add_subplot()
ax.scatter(x, y)

ax.set_title("品質モンスター散布図")
ax.set_xlabel("生産数")
ax.set_ylabel("不良数");
```
