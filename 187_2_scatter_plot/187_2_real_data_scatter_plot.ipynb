```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

anime_master_csv = "dataset/anime_master.csv"

df = pd.read_csv(anime_master_csv)
df.head()
```

```python
plt.style.use("ggplot")

members = df["members"]
rating = df["rating"]

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)

ax.scatter(members, rating, alpha=0.5)
```

```python
fig = plt.figure(figsize=(10, 5))
ax = fig.add_subplot(1, 1, 1)

types = df["type"].unique()

for t in types:
    members = df.loc[df["type"] == t, "members"]
    rating = df.loc[df["type"] == t, "rating"]
    ax.scatter(members, rating, alpha=0.5, label=t)

ax.set_xlabel("members")
ax.set_ylabel("rating")
ax.legend(types, loc="lower right")
```
