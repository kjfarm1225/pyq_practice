## 190-1の総復習問題

```python
import numpy as np
```

```python
x = np.array([[10, -20, 30], [40, 50, -60]])
print("---- x ----")
print(x)
y = np.arange(25, 55, 5).reshape((2, 3))
print("---- y ----")
print(y)
z = np.arange(10, 13, 0.5).reshape((3, 2))
print("---- z ----")
print(z)
```

```python
# 行列xの要素の絶対値
x_abs = np.abs(x)
x_abs
```

```python
# 行列yの要素の自然対数をとる
y_log = np.log(y)
y_log
```

```python
# 行列xのすべての要素に5を足す
x1 = x + 5
print(x)
x1
```

```python
# 行列xと行列yの和
print(x)
print(y)
xy = x + y
xy
```

```python
# 行列xと行列zの行列の積
prod_xz = np.dot(x, z)
print(x)
print(z)
prod_xz
```

```python
# 行列zと行列xの行列の積
prod_zx = np.dot(z, x)
print(z)
print(x)
prod_zx
```

```python
# 行列xと行列yを比較して、要素ごとに等しいかを出力する行列
copy_xy = x == y
print(x)
print(y)
copy_xy
```
