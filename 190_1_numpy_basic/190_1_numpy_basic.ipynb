## 190-1: NumPyの使い方を学ぼう 1問目

```python
import numpy as np
```

```python
# 1次元配列を作ります
a = np.array([1, 2, 3])
a
```

```python
# 2次元配列を作ります
b = np.array([[1, 2, 3], [4, 5, 6]])
b
```

```python
# 配列aの2〜3番目の要素
a[1:]
```

```python
# 行列bの1、2行目、2〜3番目の要素
b[:, 1:]
```

```python
# 行列bの1行目、2〜3番目の要素
b[:1, 1:]
```

```python
# 0から9までの数列
c = np.arange(10)
c
```

```python
# 1から10までの数列
d = np.arange(1, 11)
d
```

```python
# 引数が3つあるとき、3番目の引数は数同士の間隔になります
c2 = np.arange(1, 11, 2)
c2
```

```python
x = np.array([10, 20, 30, 40])
y = np.array([50, 60, 70, 80])
```

```python
# 配列x, yを水平方向に結合
xy_h = np.hstack([x, y])
xy_h
```

```python
# 配列x, yを垂直方向に結合
xy_v = np.vstack([x, y])
xy_v
```

```python
# 指定された範囲で均等割りした配列
z = np.linspace(0, 1, 5)
z
```
