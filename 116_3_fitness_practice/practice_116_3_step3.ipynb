```python
import pandas as pd
df = pd.read_csv("dataset/fitness_after.csv", index_col=0, parse_dates=["開始日時"])
df.head()
```

```python
df = df.set_index("開始日時")
df.head()
```

```python
summary_sr = df.resample("D")["分"].sum()
summary_df = summary_sr.to_frame()
summary_df.head()
```

```python
names = df["カテゴリー"].unique()
names
```

```python
for category in names:
    selected_df = df[df["カテゴリー"] == category]
    summary_df[category] = selected_df.resample("D")["分"].sum()
summary_df.head()
```

```python
summary_df.tail()
```

```python
summary_df.loc[summary_df["筋トレ"].isna(), "筋トレ"] = 0
summary_df.loc[summary_df["ウォーキング"].isna(), "ウォーキング"] = 0
summary_df.tail()
```

```python
def is_ok(sr):
    return (sr["ストレッチ"] >= 5) & (sr["分"] >=45)
new_sr = summary_df.apply(is_ok, axis=1)
new_sr.head()
```

```python
summary_df["デイリー目標OK"] = new_sr
summary_df.head()
```

```python
summary_df.index.name = "日付"
summary_df = summary_df.rename(columns={"分": "合計(分)", })
summary_df.head()
```
