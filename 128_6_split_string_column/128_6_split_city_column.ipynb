###### 128_6_split_string_column

```python
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        "国": ["日本", "フランス"],
        "都市": ["東京/大阪", "パリ"],
    }
)
df
```

```python
df[["都市1", "都市2"]] = df["都市"].str.extract(r"(\w+)/(\w*)")
df
```

```python
df.drop("都市", axis=1, inplace=True)
df
```
