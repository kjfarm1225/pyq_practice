```python
df_site_master = pd.DataFrame(
    [
        ["MELMB", "メルモビ", "品質保証1課", "重点監視"],
        ["WSE", "中国WSE", "品質保証2課", "重点監視"],
        ["GSE", "中国GSE", "品質保証2課", "通常監視"],
        ["ISE", "日本ISE", "品質保証1課", "通常監視"],
    ],
    columns=[
        "site",
        "site_name",
        "owner_department",
        "monitoring_level",
    ],
)
df_site_master.head()
```

```python
df_cross = df_defect.pivot_table(
    values="defect_id",
    index="site",
    columns="defect_category",
    aggfunc="count",
    fill_value=0,
)
df_cross
```

```python
df_cross = df_cross[
    [
        "Buck-IC",
        "NTF",
        "はんだ",
        "コイル",
        "異物",
    ]
]
df_cross
```

```python
df_cross["total"] = df_cross.sum(axis=1)
df_cross
```

```python
df_cross_sort = df_cross.sort_values(
    "total",
    ascending=False,
)
df_cross_sort
```

```python
df_cross_sort = df_cross_sort.reset_index()
```

```python
df_cross_sort
```

```python
df_merge = df_cross_sort.merge(
    df_site_master,
    on="site",
)
```

```python
df_merge
```

```python
df_report = df_merge[
    [
        "owner_department",
        "site_name",
        "site",
        "monitoring_level",
        "Buck-IC",
        "NTF",
        "はんだ",
        "コイル",
        "異物",
        "total",
    ]
].copy()
```

```python
df_report
df_report["check_level"] = "通常確認"
df_report

df_report.loc[
    df_report["コイル"] >= 4,
    "check_level",
] = "コイル重点確認"
df_report

df_report.loc[
    df_report["Buck-IC"] >= 3,
    "check_level",
] = "Buck-IC重点確認"
df_report

df_report.loc[
    df_report["異物"] >= 3,
    "check_level",
] = "異物重点確認"
df_report
```

```python
df_priority = df_report[
    df_report["check_level"] != "通常確認"
]
```
