```python
import pandas as pd
```

```python
df = pd.read_csv(
    "dataset/lcm_style_check_practice.csv",
)
```

```python
df
```

```python
df["defect_count_num"] = pd.to_numeric(
    df["defect_count"],
    errors="coerce",
)

df["defect_rate_num"] = pd.to_numeric(
    df["defect_rate"],
    errors="coerce",
)
df.head()
```

```python
df_defect_count_ng = df[
    df["defect_count_num"].isna()
].copy()
display(df_defect_count_ng)

df_defect_rate_ng = df[
    df["defect_rate_num"].isna()
].copy()
display(df_defect_rate_ng)
```

```python
def style_defect_count(x):
    value =  pd.to_numeric(
        x,
        errors="coerce",
    )

    if pd.isna(value):
        style = "background-color: lightgray;"
    elif value >= 5:
        style = "font-weight: bold;"
    else:
        style = None

    return style
```

```python
#cell6 
def style_defect_rate(x):
    value = pd.to_numeric(
        x,
        errors="coerce",
    )

    if pd.isna(value):
        style = "background-color: lightgray;"
    elif value >= 2.0:
        style = "color:red; font-weight:bold;"
    else:
        style = None
    return style
```

```python
def style_status(x):
    if x == "未対応":
        style = "background-color: yellow; color: black"
    elif x == "不可":
        style = "background-color: lightgray;"
    else:
        style = None

    return style
```

```python
style_df = (
    df.style
    .map(style_defect_count, subset=["defect_count"])
    .map(style_defect_rate, subset=["defect_rate"])
    .map(style_status, subset=["status"])
)

style_df
```
