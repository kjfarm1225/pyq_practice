```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
df_defect = pd.read_csv("dataset/defect_comment_reason.csv")
df_defect
```

```python
#cell4
df_defect["exclude_reason"] = ""
df_defect
```

```python
df_defect.loc[
    df_defect["comment"].str.contains(
        "TEST",
        case=False,
        na=False,
    )
    & (df_defect["exclude_reason"] == ""),
    "exclude_reason",
] = "TESTデータ"
df_defect
```

```python
#cell6

df_defect.loc[
    df_defect["comment"].str.contains(
        "確認用",
        na=False,
    )
    & (df_defect["exclude_reason"] == ""),
    "exclude_reason",
] = "確認用データ"
df_defect
```

```python
df_defect.loc[
    df_defect["comment"].str.contains(
    "サンプル",
    na=False,
    )
    & (df_defect["exclude_reason"] == ""),
    "exclude_reason"
] = "サンプル評価"
df_defect
```

```python
#cell8
df_defect.loc[
    df_defect["comment"].str.contains(
        "重複",
        na=False,
    )
    & (df_defect["exclude_reason"] == ""),
    "exclude_reason",
] = "重複登録"
df_defect
```

```python
df_defect.loc[
    df_defect["comment"].str.contains(
        "転記ミス",
        na=False,
    )
    & (df_defect["exclude_reason"] == ""),
    "exclude_reason",
] = "転記ミス"
df_defect
```

```python
df_actual = df_defect[
    df_defect["exclude_reason"] == ""
].copy()

df_actual
```

```python
df_excluded = df_defect[
    df_defect["exclude_reason"] != ""
].copy()
df_excluded
```

```python
df_site_count = df_actual.groupby(
    "site",
    as_index=False,
)["defect_id"].count()

df_site_count

df_site_count = df_site_count.rename(
    columns={
        "defect_id": "actual_count",
    }
)
df_site_count

df_site_count = df_site_count.sort_values(
    ["actual_count", "site"],
    ascending=[False, True],
).reset_index(drop=True)

df_site_count
```

```python
df_reason_count = df_excluded.groupby(
    "exclude_reason",
    as_index=False,
)["defect_id"].count()
df_reason_count

df_reason_count = df_reason_count.rename(
    columns={
        "defect_id": "count",
    }
)
df_reason_count

df_reason_count = df_reason_count.sort_values(
    ["count","exclude_reason"],
    ascending=[False, True],
).reset_index(drop=True)

df_reason_count
```
