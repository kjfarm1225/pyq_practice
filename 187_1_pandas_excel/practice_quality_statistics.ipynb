```python
import pandas as pd
from pathlib import Path

data_dir =  Path("dataset")
data_dir.mkdir(exist_ok=True)
```

```python
sample_df = pd.DataFrame({
     "month": ["1月", "2月", "3月", "4月", "5月", "6月", "1月", "2月", "3月", "4月", "5月", "6月"],
    "site": ["MELMB", "MELMB", "MELMB", "MELMB", "MELMB", "MELMB", "WSE", "WSE", "WSE", "WSE", "WSE", "WSE"],
    "production": [12000, 13500, 12800, 14000, 15000, 14800, 9000, 9500, 9800, 10200, 11000, 10800],
    "defects": [24, 32, 18, 40, 22, 15, 30, 28, 45, 20, 18, 16],
})
```

```python
quality_xlsx = "dataset/quality_data.xlsx"
```

```python
sample_df.to_excel(quality_xlsx, index=False)
```

```python
df = pd.read_excel(quality_xlsx)
```

```python
df.head()
```

```python
df["ppm"] = (df["defects"] / df["production"] * 1000000).round(1)    
```
