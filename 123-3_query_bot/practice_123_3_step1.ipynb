```python
import pandas as pd
from IPython.display import display

def query_bot(fnam, max_rows=5, **option):
    # Excelファイルを読み込む
    df = pd.read_excel(fnam, **option)

    # 名前の表記ゆれ（全角スペース）を半角スペースに直す
    df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)

    # 表示行数を一時的に制限する
    with pd.option_context("display.max_rows", max_rows):
        # 空文字が入力されるまで繰り返す
        while s := input("> "):
            try:
                # 入力された文字列を query として実行
                result = df.query(s)

                # 見やすく表示
                display(result)

            except:
                print("クエリを正しくしてください")

names = ["名前", "クラス", "クラブ", "実施月", "数学"]
query_bot("input/exam.xlsx", header=1, names=names)
```

```python
import pandas as pd
from IPython.display import display

def query_bot(fnam, max_rows=5, **option):
    df = pd.read_excel(fnam, **option)
    df["名前"] = df["名前"].str.replace("\u3000", " ", regex=False)

    print("クエリーを入力してください。空Enterで終了します。")

    with pd.option_context("display.max_rows", max_rows):
        while s := input("> "):
            try:
                result = df.query(s)
                display(result)
            except:
                print("クエリを正しくしてください")
names = ["名前", "クラス", "クラブ", "実施月", "数学"]
query_bot("input/exam.xlsx", header=1, names=names)
```
