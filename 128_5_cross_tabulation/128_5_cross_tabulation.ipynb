###### 128_5_cross_tabulation

```python
import numpy as np
import pandas as pd
```

```python
rnd = np.random.default_rng(2)
```

```python
data1 = rnd.choice(
    [
        "1-A",
        "1-B",
    ],
    60,
)
data1
```

```python
data2 = rnd.choice(
    [
        "Science",
        "Social",
        "Sport",
    ],
    60,
)
data2
```

```python
df = pd.DataFrame(
    {
        "class": data1,
        "club": data2,
    }
)
df.head()
```

```python
dfc = df.reset_index()
dfc.head()
```

```python
df_result = dfc.pivot_table(
    values="index",
    columns="club",
    index="class",
    aggfunc="count",
)
df_result
```
