```python
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv(
    "dataset/study_log_jst.csv",
    parse_dates=[
        "start_at",
        "end_at",
    ],
)
df
```

```python
df.dtypes
```

```python
df["start_date"] = df["start_at"].dt.date

df[
    [
        "start_at",
        "start_date",
    ]
]
```

```python
df["start_weekday"] = df["start_at"].dt.day_name(
    "ja_JP",
)

df[
    [
        "start_at",
        "start_weekday"
    ]
]
```
