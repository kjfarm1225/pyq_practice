```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/review.csv", index_col="user_id")
df
```

```python
contains_py = df["comment"].str.contains(
    "PY",
    case=False,
)
contains_py
```

```python
not_contains_py = ~contains_py
not_contains_py
```

```python
selected_df = df[not_contains_py].copy()
selected_df
```
