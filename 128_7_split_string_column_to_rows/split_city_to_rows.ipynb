###### 128_7_split_string_column_to_rows

```python
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        "国": [
            "日本",
            "フランス",
        ],
        "都市": [
            "東京/大阪",
            "パリ",
        ],
    }
)
df
```

```python
df = df.set_index("国").都市.str.extractall(r"(?P<都市>[^/]+)")
df = df.reset_index("国").reset_index(drop=True)
```
