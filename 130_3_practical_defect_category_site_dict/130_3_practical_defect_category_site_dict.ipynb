```python
#cell 1
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
#cell2 
df_defect = pd.DataFrame(
    [
        ["SUS", "J69P", "コイル", 3],
        ["SUS", "J69P", "Buck-IC", 1],
        ["SUS", "J69P", "コイル", 2],
        ["SUS", "310D", "NTF", 1],
        ["WSE", "310D", "コイル", 2],
        ["WSE", "310D", "NTF", 0],
        ["GSE", "J69P", "コイル", 4],
        ["GSE", "J69P", "Buck-IC", 2],
        ["MELMB", "310D", "コイル", 5],
        ["MELMB", "310D", "NTF", 1],
    ],
    columns=[
        "site",
        "model",
        "defect_category",
        "defect_qty",
    ],
)
df_defect
```

```python
#cell3 

df_work = df_defect[
    [
        "defect_category",
        "site",
        "defect_qty",
    ]
].copy()

df_work
```

```python
#cell4
df_work_positive = df_work[
    df_work["defect_qty"] > 0
].copy()

df_work_positive
```

```python
#cell5
df_unique = df_work_positive.drop_duplicates(
    [
        "defect_category",
        "site",
    ]
)
df_unique
```

```python
#cell6
sr_category_site = (
    df_unique.groupby("defect_category")["site"].apply(list)
)
sr_category_site
```

```python
#cell7
dict_category_site = sr_category_site.to_dict()
dict_category_site
```
