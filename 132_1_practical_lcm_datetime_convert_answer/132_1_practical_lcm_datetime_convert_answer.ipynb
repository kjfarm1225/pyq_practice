```python
df_defect["occ_at"] = pd.to_datetime(
    df_defect["occ_at"],
    format="mixed",
    errors="coerce",
)
df_defect
```

```python
df_defect["received_at"] = pd.to_datetime(
    df_defect["received_at"],
    format="mixed",
    errors="coerce",
)
df_defect
```

```python
df_defect["answered_at"] = pd.to_datetime(
    df_defect["answered_at"],
    format="mixed",
    errors="coerce",
)
df_defect
```

```python
df_defect.dtypes
```

```python
df_defect["occ_month"] = (
    df_defect["occ_at"]
    .dt.to_period("M")
    .astype(str)
)
df_defect.head()
```

```python
#cell 9
df_answered = df_defect[
    df_defect["answered_at"].notna()
].copy()
df_answered
```

```python
df_not_answered = df_defect[
    df_defect["answered_at"].isna()
].copy()
df_not_answered
```

```python
df_answered["response_days"] = (
    df_answered["answered_at"] - df_answered["occ_at"]
).dt.days

df_answered
```

```python
df_answered["response_rank"]  = ""
df_answered
```

```python
df_answered.loc[
    df_answered["response_days"] <= 2,
    "response_rank",
] = "早い"
df_answered
```

```python
df_answered.loc[
    (df_answered["response_days"] >= 3)
    & (df_answered["response_days"] <= 5),
    "response_rank",
] = "標準"
df_answered
```

```python
df_answered.loc[
    df_answered["response_days"] >= 6,
    "response_rank"
] = "遅い"
df_answered
```

```python
#cell 16

df_month_response = df_answered.groupby(
    "occ_month",
    as_index=False,
)["response_days"].mean()
df_month_response
```

```python
df_month_response = df_month_response.rename(
    columns={
        "response_days": "avg_response_days",
    }
)
df_month_response
```

```python
df_month_response["avg_response_days"] = (
    df_month_response["avg_response_days"].round(1)
)
df_month_response
```

```python
df_site_response = df_answered.groupby(
    "site",
    as_index=False,
)["response_days"].mean()
df_site_response
```

```python
df_site_response = df_site_response.rename(
    columns={
        "response_days": "avg_response_days",
    }
)
df_site_response
```

```python
df_site_response["avg_response_days"] = (
    df_site_response["avg_response_days"].round(1)
)
df_site_response
```

```python
df_site_response = df_site_response.sort_values(
    ["avg_response_days", "site"],
    ascending=[False, True],
).reset_index(drop=True)
df_site_response
```
