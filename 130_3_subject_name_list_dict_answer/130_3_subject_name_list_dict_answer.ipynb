```python
#cell 1
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
#cell 2
df = pd.DataFrame(
    [
        ["Alice", "国語", 87],
        ["Alice", "数学", 72],
        ["Bob", "国語", 65],
        ["Bob", "数学", 92],
        ["Carol", "数学", 98],
    ],
    columns=[
        "Name",
        "Subject",
        "Point",
    ],
)
df
```

```python
sr_name_list = df.groupby("Subject")["Name"].apply(list)
sr_name_list
```

```python
dict_result = sr_name_list.to_dict()
dict_result
```
