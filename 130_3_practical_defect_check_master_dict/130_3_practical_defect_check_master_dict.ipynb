```python
#cell1
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
#cell 2

df_defect = pd.DataFrame(
    [
        ["D001", "SUS", "J69P", "コイル", 3],
        ["D002", "SUS", "J69P", "Buck-IC", 1],
        ["D003", "WSE", "310D", "コイル", 2],
        ["D004", "GSE", "J69P", "Buck-IC", 2],
        ["D005", "MELMB", "310D", "NTF", 1],
        ["D006", "ISE", "J69P", "CAN通信", 1],
        ["D007", "SUS", "310D", "NTF", 0],
        ["D008", "MELMB", "310D", "コイル", 5],
        ["D009", "GSE", "J69P", "コイル", 1],
    ],
    columns=[
        "defect_id",
        "site",
        "model",
        "defect_category",
        "defect_qty",
    ],
)

df_defect
```

```python
df_check_master = pd.DataFrame(
    [
        ["コイル", "断面観察"],
        ["コイル", "インパルス検査条件確認"],
        ["コイル", "巻線状態・鉄粉混入確認"],
        ["Buck-IC", "外観焼損確認"],
        ["Buck-IC", "Lot / DateCode確認"],
        ["Buck-IC", "メーカー解析依頼状況確認"],
        ["NTF", "再現試験条件確認"],
        ["NTF", "車両側・単品側の切り分け確認"],
    ],
    columns=[
        "defect_category",
        "check_item",
    ],
)
df_check_master
```

```python
#cell4
df_owner_master = pd.DataFrame(
    [
        ["コイル", "部品/サプライヤ確認", "高"],
        ["Buck-IC", "電子部品解析", "高"],
        ["NTF", "再現確認チーム", "中"],
    ],
    columns=[
        "defect_category",
        "owner_team",
        "priority"
    ],
)
df_owner_master
```

```python
df_defect_positive = df_defect[
    df_defect["defect_qty"] > 0
].copy()

df_defect_positive
```

```python
#cell 6
dict_check_items = (
    df_check_master.groupby("defect_category")["check_item"]
    .apply(list)
    .to_dict()
)
dict_check_items
```

```python
#cell7

dict_owner_team = (
    df_owner_master.set_index("defect_category")["owner_team"]
    .to_dict()
)
dict_owner_team
```

```python
#cell3
dict_priority = (
    df_owner_master.set_index("defect_category")["priority"]
    .to_dict()
)
dict_priority
```

```python
#cell9
df_defect_positive["check_item_list"] = (
    df_defect_positive["defect_category"].map(dict_check_items)
)
df_defect_positive
```

```python
#cell10
df_defect_positive["owner_team"] = (
    df_defect_positive["defect_category"].map(dict_owner_team)
)
df_defect_positive
```

```python
#cell 12
df_defect_positive["check_status"] = "登録あり"
df_defect_positive.head()

df_defect_positive.loc[
    df_defect_positive["check_item_list"].isna(),
    "check_status",
] = "マスター未登録"
df_defect_positive
```

```python
df_unregistered = df_defect_positive[
    df_defect_positive["check_status"] == "マスター未登録"
].copy()

df_unregistered

```

```python
#cell 14
df_investigation_list = df_defect_positive[
    [
        "defect_id",
        "site",
        "model",
        "defect_category",
        "defect_qty",
        "owner_team",
        "priority",
        "check_status",
        "check_item_list",
    ]
].copy()
df_investigation_list
```
