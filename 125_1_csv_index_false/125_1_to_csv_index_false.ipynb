## 演習問題

```python
import pandas as pd

df_quality = pd.DataFrame(
    [
        ["melmb", "点灯不良", 10000, 3],
        ["wse", "外観不良",  8000, 5],
        ["gse", "ショート",  12000, 1],
        ["ise", "コネクタ浮き",  9000, 2],
    ],
    columns=["site", "defect", "production", "defects"],
)
```

```python
df_quality["ppm"] = df_quality["defects"] / df_quality["production"] * 1000000
df_quality["ppm"] = df_quality["ppm"].round(1)
df_quality.to_csv("quality_result.csv", index=False)
```
