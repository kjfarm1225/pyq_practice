## 実務練習問題

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(
    [
      ["2026/06/01", "MELMB", "点灯不良", 10000, 3],
        ["2026/06/01", "WSE", "外観不良", 8000, 5],
        ["2026/06/02", "GSE", "ショート", 12000, 1],
        ["2026/06/02", "MELMB", "点灯不良", 11000, 8],
        ["2026/06/03", "ISE", "コネクタ浮き", 9000, 2],
        ["2026/06/03", "WSE", "異物混入", 7000, 7],  
    ],
    columns=["data", "site", "defect", "production", "defects"],
)
```

```python
df["ppm"] = (df["defects"] / df["production"] * 1000000).round(1)
df["判定"] = np.where(df["ppm"] >= 800, "危険モンスター", "通常モンスター")

df.to_csv("output/quality_monster_report.csv", index=False)
df_sorted = df.sort_values(by="ppm", ascending=False)
df_sorted.to_csv("output/quality_monster_report_sorted.csv", index=False)
```

## 完成版コード全部入り

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(
    [
         ["2026/06/01", "MELMB", "点灯不良", 10000, 3],
        ["2026/06/01", "WSE", "外観不良", 8000, 5],
        ["2026/06/02", "GSE", "ショート", 12000, 1],
        ["2026/06/02", "MELMB", "点灯不良", 11000, 8],
        ["2026/06/03", "ISE", "コネクタ浮き", 9000, 2],
        ["2026/06/03", "WSE", "異物混入", 7000, 7],
    ],
    columns=["date", "site", "defect", "production", "defects"],
)

```

```python
df["ppm"] = (df["defects"] / df["production"] * 1000000).round(1)
df["判定"] = np.where(df["ppm"] >= 800, "危険モンスター", "通常モンスター")

df.to_csv("output/quality_monster_report.csv", index=False)
print(df)
df_sorted = df.sort_values(by="ppm", ascending=False)
df_sorted.to_csv("output/quality_monster_report_sorted.csv", index=False)
print(df_sorted)

df_simple = df_sorted[["date", "site", "defect", "ppm", "判定"]]
df_simple.to_csv("output/quality_monster_report_simple.csv", index=False)
print(df_simple)
```
