## 「140-1: Matplotlibの基本」 2問目

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x_months = [1, 2, 3, 4, 5, 6]
y_temps = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]
```

```python
plt.rcParams["font.family"] = "Meiryo"
```

```python
fig, ax = plt.subplots()
ax.plot(x_months, y_temps)
fig.suptitle("月別の最高気温", size=20)
ax.set_xlabel("月", size=15)
ax.set_ylabel("気温", size=15)
plt.show()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

x_months = [1, 2, 3, 4, 5, 6]
y_temps = [12.5, 15.3, 16.3, 22.9, 26.7, 28.4]

plt.rcParams["font.family"] = "Meiryo"

fig, ax = plt.subplots()
ax.plot(x_months, y_temps)

fig.suptitle("月別の最高気温", size=20)
ax.set_xlabel("月", size=15)
ax.set_ylabel("気温", size=15)

plt.show()
```
