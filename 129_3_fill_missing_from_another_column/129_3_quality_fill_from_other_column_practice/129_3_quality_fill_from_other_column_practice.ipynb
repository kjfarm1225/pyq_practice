```python
#cell1
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        ["D001", "J69P", "MELMB", "L250601-A", "コイル", 1, np.nan, 300],
        ["D002", "J69P", "WSE", "L250601-B", "Buck-IC", 2, 2, 280],
        ["D003", "310D", "GSE", "L250602-C", "はんだ", 1, np.nan, 250],
        ["D004", "J69P", "MELMB", "L250603-D", "コイル", 3, 4, 310],
        ["D005", "310D", "WSE", "L250604-E", "NTF", 0, np.nan, 260],
        ["D006", "J69P", "GSE", "L250605-F", "コイル", 2, np.nan, 290],
        ["D007", "310D", "MELMB", "L250606-G", "はんだ", 1, 1, 240],
        ["D008", "J69P", "WSE", "L250607-H", "Buck-IC", 1, np.nan, 270],
    ],
    columns=[
        "defect_id",
        "project",
        "site",
        "lot_no",
        "defect_category",
        "first_report_defect_qty",
        "confirmed_defect_qty",
        "inspection_qty",
    ],
)
df
```

```python
#cell3
df_work = df.copy()
df_work
```

```python
df_work["confirmed_defect_qty_filled"] = df_work["confirmed_defect_qty"].fillna(
    df_work["first_report_defect_qty"]
)
df_work
```

```python
#cell5
df_check_fill = df_work[
    [
        "defect_id",
        "project",
        "site",
        "defect_category",
        "first_report_defect_qty",
        "confirmed_defect_qty",
        "confirmed_defect_qty_filled",
    ]
].copy()

df_check_fill
```

```python
df_work["defect_rate"] = (
    df_work["confirmed_defect_qty_filled"] / df_work["inspection_qty"]
)
df_work
```

```python
df_rate_sort = df_work.sort_values(
    "defect_rate",
    ascending=False,
)
df_rate_sort
```

```python
#cell7

df_rate_sort = df_work.sort_values(
    "defect_rate",
    ascending=False,
)
df_rate_sort
```

```python
df_work["check_result"] = "通常"

df_work.loc[
    df_work["defect_rate"] > 0.005,
    "check_result",
] = "要確認"

df_work
```

```python
df_need_check = df_work[
    df_work["check_result"] == "要確認"
].copy()
df_need_check
```
