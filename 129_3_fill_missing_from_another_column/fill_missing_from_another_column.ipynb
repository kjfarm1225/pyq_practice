###### 129_3_fill_missing_from_another_column

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        ["Alice", "", 87, 83],
        ["Alice", "", 72, np.nan],
        ["Bob", "", 65, np.nan],
        ["Bob", "", 92, 86],
    ],
    columns=["name", "subject", "point", "last"],
)
df
```

```python
df["last"] = df["last"].fillna(df["point"])
df
```
