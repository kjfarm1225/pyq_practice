```python
import numpy as np
```

```python
prices = np.array([100, 130, 180])
prices
```

```python
discounts = np.array([10, 20, 25])
discounts
```

```python
quantities = np.array([20, 10, 5])
quantities
```

```python
discounted_prices = prices - discounts
discounted_prices
```

```python
amounts = discounted_prices * quantities
amounts
```
