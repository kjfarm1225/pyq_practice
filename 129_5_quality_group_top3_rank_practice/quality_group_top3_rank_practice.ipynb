```python
#cell 1
import pandas as pd
```

```python
#cell 2
df = pd.DataFrame(
    [
        ["D001", "J69P", "MELMB", "コイル", 3, "調査中"],
        ["D002", "J69P", "WSE", "Buck-IC", 5, "対策検討中"],
        ["D003", "310D", "GSE", "コイル", 12, "暫定対策済"],
        ["D004", "310D", "MELMB", "はんだ", 2, "調査中"],
        ["D005", "J69P", "GSE", "コイル", 7, "調査中"],
        ["D006", "310D", "WSE", "コネクタ", 1, "完了"],
        ["D007", "J69P", "MELMB", "Buck-IC", 8, "調査中"],
        ["D008", "310D", "GSE", "コイル", 4, "対策検討中"],
        ["D009", "J69P", "GSE", "Buck-IC", 6, "調査中"],
        ["D010", "310D", "MELMB", "コイル", 5, "暫定対策済"],
        ["D011", "J69P", "WSE", "コイル", 9, "調査中"],
        ["D012", "310D", "GSE", "はんだ", 3, "完了"],
    ],
    columns=[
        "defect_id",
        "project",
        "site",
        "defect_category",
        "defect_qty",
        "status",
    ],
)
df
```

```python
#cell3
df_work = df.copy()
df_work
```

```python
#cell 4
df_summary = df_work.groupby(
    [
        "site",
        "defect_category",
    ],
    as_index=False,
)["defect_qty"].sum()
df_summary
```

```python
#cell 5
df_summary = df_summary.rename(
    columns={
        "defect_qty": "total_defect_qty",
    }
)
df_summary
```

```python
#cell 6
top3_index = df_summary["total_defect_qty"].nlargest(3).index
top3_index
```

```python
#cell 7
df_summary.loc[
    top3_index,
    "priority_rank",
] = [
    1,
    2,
    3,
]
df_summary
```

```python
#cell 8
df_summary["priority_flag"] = "通常確認"

df_summary.loc[
    df_summary["priority_rank"].notna(),
    "priority_flag",
] = "重点確認"

df_summary
```

```python
#cell9

df_summary_sort = df_summary.sort_values(
    [
        "priority_rank",
        "total_defect_qty",
    ],
    ascending=[
        True,
        False,
    ],
)

df_summary_sort
```

```python
#cell10
df_priority = df_summary_sort[
    df_summary_sort["priority_flag"] == "重点確認"
].copy()
df_priority
```

```python
#cell 11
df_count_summary = pd.DataFrame(
    [
        ["集計後の組み合わせ数", len(df_summary)],
        ["重点確認件数", len(df_priority)],
        ["通常確認件数", len(df_summary) - len(df_priority)],
    ],
    columns=[
        "確認項目",
        "件数",
    ],
)
df_count_summary
```
